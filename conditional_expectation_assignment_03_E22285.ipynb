#Q1) Prediction of fraudulent transactions

A bank monitors credit-card transactions for possible fraud.

For each transaction, let the sample space be

$$
\Omega=\{\omega_1,\omega_2,\omega_3,\omega_4\},
$$

where:

* $\omega_1$: low-risk customer, legitimate transaction,
* $\omega_2$: low-risk customer, fraudulent transaction,
* $\omega_3$: high-risk customer, legitimate transaction,
* $\omega_4$: high-risk customer, fraudulent transaction.

Assume the probabilities are

$$
P(\{\omega_1\})=0.50,\qquad
P(\{\omega_2\})=0.10,\qquad
P(\{\omega_3\})=0.20,\qquad
P(\{\omega_4\})=0.20.
$$

Let $X:\Omega\to\mathbb R$ denote the fraud indicator:

$$
X(\omega_1)=0,\qquad
X(\omega_2)=1,\qquad
X(\omega_3)=0,\qquad
X(\omega_4)=1.
$$

Thus $X=1$ means the transaction is fraudulent, and $X=0$ means it is legitimate.

Now suppose the bank does not observe the full outcome $\omega$, but only the customer's risk class. Define $Y:\Omega\to\mathbb R$ by

$$
Y(\omega_1)=0,\qquad
Y(\omega_2)=0,\qquad
Y(\omega_3)=1,\qquad
Y(\omega_4)=1.
$$

Thus:

* $Y=0$ means “low-risk customer,”
* $Y=1$ means “high-risk customer.”

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

3. Explain why $E[X\mid Y]$ must be $\sigma(Y)$-measurable, and therefore must be constant on the sets $\{ \omega_1,\omega_2\}$ and $\{\omega_3,\omega_4\}$.

4. Compute $E[X\mid Y]$ explicitly and show that

$$
E[X\mid Y]= \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}
$$

5. Verify the defining property of conditional expectation by checking that for every set $A\in\sigma(Y)$,

$$
\int_A E[X\mid Y]\,dP=\int_A X\,dP.
$$

6. Find the Conditional PMF $p_{X|Y}(x \mid y)$.

7. Verify the identity:
$$\mathbb{E}[X\mid Y=y] = \sum_{x \in \{0,1\}} x \cdot p_{X|Y}(x \mid y)$$.

8. Calculate the **Mean Square Error (MSE)**

9. How would the **Mean Square Error (MSE)** change if the bank decided to ignore the customer risk class $Y$ and instead used the global average fraud rate ($E[X] = 0.30$) as a constant estimate for every transaction? Calculate this error and explain its relationship to the **Law of Total Variance**.

10. Using the joint distribution of $(X, Y)$ derived from the sample space:

    a) Compute the entropy $H(X)$, representing the bank's initial uncertainty regarding fraud.

    b) Compute the conditional entropy $H(X \mid Y)$, representing the uncertainty remaining after the customer's risk class is revealed.

    c) Calculate the Mutual Information $I(X; Y)$.

    d) Based on these values, what percentage of the total uncertainty about fraud is removed by knowing the customer's risk class?

## Q1.8

The Mean Square Error when using \(E[X\mid Y]\) as the estimator is

$$
\text{MSE}
=
\mathbb{E}\left[(X-E[X\mid Y])^2\right].
$$

This is also equal to

$$
\text{MSE}
=
\mathbb{E}[\mathrm{Var}(X\mid Y)].
$$

Since \(X\) is a Bernoulli random variable given \(Y=y\), we use

$$
\mathrm{Var}(X\mid Y=y)=p(1-p).
$$

For \(Y=0\),

$$
\mathrm{Var}(X\mid Y=0)
=
\frac{1}{6}\left(1-\frac{1}{6}\right)
=
\frac{1}{6}\cdot\frac{5}{6}
=
\frac{5}{36}.
$$

For \(Y=1\),

$$
\mathrm{Var}(X\mid Y=1)
=
\frac{1}{2}\left(1-\frac{1}{2}\right)
=
\frac{1}{2}\cdot\frac{1}{2}
=
\frac{1}{4}.
$$

Now,

$$
P(Y=0)=0.60=\frac{3}{5}
$$

and

$$
P(Y=1)=0.40=\frac{2}{5}.
$$

Therefore,

$$
\begin{aligned}
\mathbb{E}[\mathrm{Var}(X\mid Y)]
&=
\mathrm{Var}(X\mid Y=0)P(Y=0)
+
\mathrm{Var}(X\mid Y=1)P(Y=1)\\
&=
\left(\frac{5}{36}\cdot\frac{3}{5}\right)
+
\left(\frac{1}{4}\cdot\frac{2}{5}\right)\\
&=
\frac{1}{12}+\frac{1}{10}\\
&=
\frac{5}{60}+\frac{6}{60}\\
&=
\frac{11}{60}.
\end{aligned}
$$

Hence,

$$
\text{MSE}
=
\frac{11}{60}
\approx 0.1833.
$$

Therefore, the minimum mean square error is

$$
\boxed{\frac{11}{60}\approx 0.1833}.
$$

# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

1. Show that the conditional distribution is given by $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
* $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

## Q2

To find the conditional density of \(\mathbf{X}\) given \(\mathbf{Y}=\mathbf{y}\), we use

$$
f_{\mathbf{X}\mid \mathbf{Y}}(\mathbf{x}\mid \mathbf{y})
=
\frac{f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})}
{f_{\mathbf{Y}}(\mathbf{y})}.
$$

Let

$$
\widetilde{\mathbf{x}}=\mathbf{x}-\boldsymbol{\mu}_X
$$

and

$$
\widetilde{\mathbf{y}}=\mathbf{y}-\boldsymbol{\mu}_Y.
$$

The covariance matrix is partitioned as

$$
\boldsymbol{\Sigma}
=
\begin{bmatrix}
\boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY}\\
\boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY}
\end{bmatrix}.
$$

Using the Schur complement of \(\boldsymbol{\Sigma}_{YY}\), define

$$
\mathbf{S}_{XX}
=
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}
\boldsymbol{\Sigma}_{YY}^{-1}
\boldsymbol{\Sigma}_{YX}.
$$

The joint quadratic form can be written as

$$
Q_{\text{joint}}
=
(\mathbf{y}-\boldsymbol{\mu}_Y)^T
\boldsymbol{\Sigma}_{YY}^{-1}
(\mathbf{y}-\boldsymbol{\mu}_Y)
+
(\mathbf{x}-\boldsymbol{\mu}^*)^T
\mathbf{S}_{XX}^{-1}
(\mathbf{x}-\boldsymbol{\mu}^*),
$$

where

$$
\boldsymbol{\mu}^*
=
\boldsymbol{\mu}_X
+
\boldsymbol{\Sigma}_{XY}
\boldsymbol{\Sigma}_{YY}^{-1}
(\mathbf{y}-\boldsymbol{\mu}_Y).
$$

The marginal density of \(\mathbf{Y}\) contains the quadratic term

$$
Q_{\text{marginal}}
=
(\mathbf{y}-\boldsymbol{\mu}_Y)^T
\boldsymbol{\Sigma}_{YY}^{-1}
(\mathbf{y}-\boldsymbol{\mu}_Y).
$$

When forming the conditional density,

$$
f_{\mathbf{X}\mid \mathbf{Y}}(\mathbf{x}\mid \mathbf{y})
=
\frac{f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})}
{f_{\mathbf{Y}}(\mathbf{y})},
$$

the marginal term involving only \(\mathbf{y}\) cancels from the exponent. Therefore, the remaining quadratic term is

$$
Q_{\text{conditional}}
=
(\mathbf{x}-\boldsymbol{\mu}^*)^T
\mathbf{S}_{XX}^{-1}
(\mathbf{x}-\boldsymbol{\mu}^*).
$$

Now consider the normalizing constants. The joint density has constant

$$
\frac{1}{\sqrt{(2\pi)^{n+m}|\boldsymbol{\Sigma}|}},
$$

and the marginal density of \(\mathbf{Y}\) has constant

$$
\frac{1}{\sqrt{(2\pi)^m|\boldsymbol{\Sigma}_{YY}|}}.
$$

Dividing these constants gives

$$
\frac{1}{\sqrt{(2\pi)^{n+m}|\boldsymbol{\Sigma}|}}
\sqrt{(2\pi)^m|\boldsymbol{\Sigma}_{YY}|}
=
\frac{1}{\sqrt{(2\pi)^n
\frac{|\boldsymbol{\Sigma}|}{|\boldsymbol{\Sigma}_{YY}|}}}.
$$

Using the determinant identity

$$
|\boldsymbol{\Sigma}|
=
|\boldsymbol{\Sigma}_{YY}|
|\mathbf{S}_{XX}|,
$$

we get

$$
\frac{|\boldsymbol{\Sigma}|}{|\boldsymbol{\Sigma}_{YY}|}
=
|\mathbf{S}_{XX}|.
$$

Therefore, the conditional density becomes

$$
f_{\mathbf{X}\mid \mathbf{Y}}(\mathbf{x}\mid \mathbf{y})
=
\frac{1}{\sqrt{(2\pi)^n|\mathbf{S}_{XX}|}}
\exp\left(
-\frac{1}{2}
(\mathbf{x}-\boldsymbol{\mu}^*)^T
\mathbf{S}_{XX}^{-1}
(\mathbf{x}-\boldsymbol{\mu}^*)
\right).
$$

Hence,

$$
(\mathbf{X}\mid \mathbf{Y}=\mathbf{y})
\sim
\mathcal{N}
\left(
\boldsymbol{\mu}^*,
\mathbf{S}_{XX}
\right),
$$

where

$$
\boldsymbol{\mu}^*
=
\boldsymbol{\mu}_X
+
\boldsymbol{\Sigma}_{XY}
\boldsymbol{\Sigma}_{YY}^{-1}
(\mathbf{y}-\boldsymbol{\mu}_Y),
$$

and

$$
\mathbf{S}_{XX}
=
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}
\boldsymbol{\Sigma}_{YY}^{-1}
\boldsymbol{\Sigma}_{YX}.
$$

Therefore,

$$
\boxed{
\boldsymbol{\Sigma}_{X\mid Y}
=
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}
\boldsymbol{\Sigma}_{YY}^{-1}
\boldsymbol{\Sigma}_{YX}
}
$$

and

$$
\boxed{
\boldsymbol{\mu}_{X\mid Y}
=
\boldsymbol{\mu}_X
+
\boldsymbol{\Sigma}_{XY}
\boldsymbol{\Sigma}_{YY}^{-1}
(\mathbf{y}-\boldsymbol{\mu}_Y)
}.
$$

# Q3) Continuous Fault Monitoring of a Spindle Motor

A factory monitors the health of a high-speed CNC spindle. The critical variable is the **Internal Temperature ($X$)**, but because the motor is sealed, they can only observe the **Total Spindle Power ($Y$)**.

**The Setup:**
1.  **Hidden Temperature ($X$):** The standardized temperature deviation follows a standard normal distribution:
    $$X \sim \mathscr{N}(0, 1)$$
2.  **Mechanical Load ($Q$):** The torque required to cut material creates "noise" in the power reading. It is independent of temperature and follows:
    $$Q \sim \mathscr{N}(0, 4)$$
3.  **The Measurement ($Y$):** The power sensor $Y$ measures the combined effect of thermal load and mechanical torque:
$$Y = X + Q$$

---

**Answer the followings:**

**1. Joint Distribution**
Identify the distribution of the random varaible $(X, Y)$.
* Show that $Y \sim \mathscr{N}(0, 5)$.
* Find the covariance $\text{Cov}(X, Y)$.

**2. Information Geometry**
Explain why $X$ is **not** $\sigma(Y)$-measurable. In practical terms, if the power meter shows $Y = 5$, why can't the engineer know for certain if the motor is overheating ($X$ is high) or if the material being cut is simply very hard ($Q$ is high)?

**3. Conditional Expectation (The Best Estimate)**
Compute the conditional expectation $E[X \mid Y]$.
* Using the property for joint normals: $E[X \mid Y = y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y)$.
* Express this as a "signal-to-total-variance" ratio.

**4. Verification of the Residual**
Let $\widehat{X} = E[X \mid Y]$. Define the estimation error (residual) as $Z = X - \widehat{X}$.
* Prove that $E[Z] = 0$.
* Verify that the error $Z$ is uncorrelated with the observation $Y$ (the Orthogonality Principle).

**5. Information Theory: Differential Entropy**
* Compute the differential entropy $h(X)$.
* Compute the conditional differential entropy $h(X \mid Y)$.
* *Note: The variance of the conditional distribution $(X \mid Y=y)$ is given by $\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)}$.*

**6. Mutual Information**
Calculate $I(X; Y) = h(X) - h(X \mid Y)$.
* What percentage of the "uncertainty" about the temperature is removed by observing the power meter?
* How would this change if the mechanical noise $Q$ had a variance of $100$ instead of $4$?

**7. Diagnostic Interpretation**
The factory manager says: *"Since $Y = X + Q$, our best estimate of temperature should just be $Y$ itself."*
Using your result from Question 3, explain why $Y$ is a **biased** estimator of $X$ and why $E[X \mid Y]$ "shrinks" the measurement toward the mean (zero).



## Q3.1

Since \(X\) and \(Q\) are independent normal random variables, their sum is also normally distributed. Therefore,

$$
Y=X+Q
$$

is normally distributed.

The mean of \(Y\) is

$$
E[Y]=E[X]+E[Q].
$$

So,

$$
E[Y]=0+0=0.
$$

Because \(X\) and \(Q\) are independent,

$$
\mathrm{Var}(Y)=\mathrm{Var}(X)+\mathrm{Var}(Q).
$$

Thus,

$$
\mathrm{Var}(Y)=1+4=5.
$$

Therefore,

$$
Y\sim \mathcal{N}(0,5).
$$

Now we find the covariance between \(X\) and \(Y\):

$$
\mathrm{Cov}(X,Y)=E[XY]-E[X]E[Y].
$$

Since \(E[X]=0\) and \(E[Y]=0\),

$$
\mathrm{Cov}(X,Y)=E[XY].
$$

Using \(Y=X+Q\),

$$
E[XY]=E[X(X+Q)].
$$

Therefore,

$$
E[XY]=E[X^2]+E[XQ].
$$

Since \(X\) and \(Q\) are independent,

$$
E[XQ]=E[X]E[Q]=0.
$$

Also,

$$
E[X^2]=\mathrm{Var}(X)=1.
$$

Hence,

$$
\mathrm{Cov}(X,Y)=1.
$$

So the joint distribution of \((X,Y)\) is

$$
\begin{pmatrix}
X\\
Y
\end{pmatrix}
\sim
\mathcal{N}
\left(
\begin{pmatrix}
0\\
0
\end{pmatrix},
\begin{pmatrix}
1 & 1\\
1 & 5
\end{pmatrix}
\right).
$$

## Q3.2

The random variable \(X\) is not \(\sigma(Y)\)-measurable because \(X\) cannot be determined exactly from \(Y\).

That means there is no deterministic function \(g\) such that

$$
X=g(Y).
$$

Since

$$
Y=X+Q,
$$

a fixed observed value \(Y=y\) can come from many possible pairs \((x,q)\) satisfying

$$
x+q=y.
$$

For example, if

$$
Y=5,
$$

the engineer cannot know with certainty whether the large value is due to high internal temperature \(X\), or due to high mechanical load \(Q\).

Therefore, \(Y\) gives information about \(X\), but it does not fully reveal \(X\).

## Q3.3

For jointly normal random variables, the conditional expectation is

$$
E[X\mid Y=y]
=
\mu_X+
\frac{\mathrm{Cov}(X,Y)}{\mathrm{Var}(Y)}
(y-\mu_Y).
$$

Here,

$$
\mu_X=0,
\qquad
\mu_Y=0,
\qquad
\mathrm{Cov}(X,Y)=1,
\qquad
\mathrm{Var}(Y)=5.
$$

Substituting these values,

$$
E[X\mid Y=y]
=
0+\frac{1}{5}(y-0).
$$

Therefore,

$$
E[X\mid Y=y]=\frac{1}{5}y.
$$

Hence,

$$
E[X\mid Y]=\frac{1}{5}Y.
$$

So,

$$
\boxed{E[X\mid Y]=0.2Y}.
$$

The coefficient

$$
\frac{\mathrm{Cov}(X,Y)}{\mathrm{Var}(Y)}
=
\frac{1}{5}
$$

is the signal-to-total-variance ratio. Since only \(1\) out of the total variance \(5\) comes from \(X\), only \(20\%\) of the observed power reading is used as the temperature estimate.

## Q3.4

Let

$$
\widehat{X}=E[X\mid Y].
$$

From Q3.3,

$$
\widehat{X}=0.2Y.
$$

The residual is

$$
Z=X-\widehat{X}.
$$

Therefore,

$$
Z=X-0.2Y.
$$

First, we calculate \(E[Z]\):

$$
E[Z]=E[X-0.2Y].
$$

So,

$$
E[Z]=E[X]-0.2E[Y].
$$

Since

$$
E[X]=0
$$

and

$$
E[Y]=0,
$$

we get

$$
E[Z]=0-0.2(0)=0.
$$

Hence,

$$
\boxed{E[Z]=0}.
$$

Now we check whether \(Z\) is uncorrelated with \(Y\).

Since both \(Z\) and \(Y\) have mean zero,

$$
\mathrm{Cov}(Z,Y)=E[ZY].
$$

Now,

$$
\mathrm{Cov}(Z,Y)
=
E[(X-0.2Y)Y].
$$

Expanding,

$$
\mathrm{Cov}(Z,Y)
=
E[XY]-0.2E[Y^2].
$$

Here,

$$
E[XY]=1
$$

and

$$
E[Y^2]=\mathrm{Var}(Y)=5.
$$

Therefore,

$$
\mathrm{Cov}(Z,Y)=1-0.2(5).
$$

So,

$$
\mathrm{Cov}(Z,Y)=1-1=0.
$$

Hence,

$$
\boxed{\mathrm{Cov}(Z,Y)=0}.
$$

Therefore, the residual \(Z\) is uncorrelated with the observation \(Y\). This verifies the orthogonality principle.

## Q3.7

The manager suggests using

$$
\widehat{X}=Y.
$$

However, this is a naive estimator because \(Y\) contains both the temperature \(X\) and the mechanical noise \(Q\).

Since

$$
Y=X+Q,
$$

using \(Y\) directly also includes the noise term \(Q\).

If we use

$$
\widehat{X}=Y,
$$

then the error is

$$
X-Y.
$$

Since

$$
Y=X+Q,
$$

we get

$$
X-Y=X-(X+Q).
$$

Therefore,

$$
X-Y=-Q.
$$

The error variance is

$$
E[(X-Y)^2]=E[Q^2].
$$

Since

$$
\mathrm{Var}(Q)=4,
$$

we get

$$
E[(X-Y)^2]=4.
$$

But using the conditional expectation,

$$
E[X\mid Y]=0.2Y.
$$

The error variance is only

$$
0.8.
$$

Therefore, \(E[X\mid Y]\) gives a much smaller error than using \(Y\) directly.

The factor \(0.2\) shows shrinkage. It pulls the estimate toward the mean value \(0\), instead of fully trusting the noisy observation \(Y\).

So, when \(Y\) is large, the best estimate does not assume that all of the increase is due to temperature. Most of the variation may come from the mechanical load \(Q\).

Hence,

$$
\boxed{E[X\mid Y]=0.2Y}
$$

is the better estimate of the internal temperature.

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from scipy.stats import multivariate_normal

# -----------------------------
# Parameters of the spindle model
# -----------------------------
# X = internal temperature deviation
# Q = mechanical load noise
# Y = observed power reading = X + Q

mean_x = 0
mean_q = 0

var_x = 1
var_q = 4

mean_y = mean_x + mean_q
var_y = var_x + var_q

cov_xy = var_x
gain = cov_xy / var_y   # gain = 1/5 = 0.2

# -----------------------------
# Generate random samples
# -----------------------------
np.random.seed(7)

sample_size = 1000

temperature_x = np.random.normal(mean_x, np.sqrt(var_x), sample_size)
load_q = np.random.normal(mean_q, np.sqrt(var_q), sample_size)

power_y = temperature_x + load_q

estimated_x = gain * power_y
error_z = temperature_x - estimated_x

# -----------------------------
# Create grid for joint density
# -----------------------------
x_values = np.linspace(-4, 4, 200)
y_values = np.linspace(-8, 8, 200)

X_grid, Y_grid = np.meshgrid(x_values, y_values)

grid_points = np.dstack((X_grid, Y_grid))

mean_vector = [mean_x, mean_y]
cov_matrix = [
    [var_x, cov_xy],
    [cov_xy, var_y]
]

density = multivariate_normal(mean_vector, cov_matrix).pdf(grid_points)

# -----------------------------
# Create plots
# -----------------------------
fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=(
        "Joint Behaviour of Temperature and Power<br><sup>Optimal Estimate Compared with Naive Estimate</sup>",
        "Residual Error Against Power Reading<br><sup>Checking Orthogonality</sup>"
    )
)

# Left plot: joint distribution
fig.add_trace(
    go.Contour(
        x=x_values,
        y=y_values,
        z=density,
        colorscale="Blues",
        opacity=0.65,
        showscale=False,
        hoverinfo="skip",
        name="Joint Density"
    ),
    row=1,
    col=1
)

fig.add_trace(
    go.Scatter(
        x=temperature_x,
        y=power_y,
        mode="markers",
        marker=dict(size=3, color="rgba(0,0,0,0.18)"),
        name="Simulated Samples"
    ),
    row=1,
    col=1
)

line_y = np.linspace(-8, 8, 100)

fig.add_trace(
    go.Scatter(
        x=gain * line_y,
        y=line_y,
        mode="lines",
        line=dict(width=4, color="red"),
        name="Optimal Estimate: E[X|Y] = 0.2Y"
    ),
    row=1,
    col=1
)

fig.add_trace(
    go.Scatter(
        x=line_y,
        y=line_y,
        mode="lines",
        line=dict(width=2, color="black", dash="dash"),
        name="Naive Estimate: X = Y"
    ),
    row=1,
    col=1
)

# Right plot: residual against Y
fig.add_trace(
    go.Scatter(
        x=power_y,
        y=error_z,
        mode="markers",
        marker=dict(size=4, color="rgba(120,0,160,0.45)"),
        name="Residual: Z = X - 0.2Y"
    ),
    row=1,
    col=2
)

fig.add_trace(
    go.Scatter(
        x=line_y,
        y=np.zeros_like(line_y),
        mode="lines",
        line=dict(width=2, color="black"),
        name="Zero Error Line"
    ),
    row=1,
    col=2
)

# -----------------------------
# Layout settings
# -----------------------------
fig.update_layout(
    title="Spindle Motor Monitoring: Estimating Hidden Temperature from Noisy Power Reading",
    height=600,
    template="plotly_white",
    hovermode="closest"
)

fig.update_xaxes(title_text="Internal Temperature (X)", row=1, col=1)
fig.update_yaxes(title_text="Observed Power (Y)", row=1, col=1)

fig.update_xaxes(title_text="Observed Power (Y)", row=1, col=2)
fig.update_yaxes(title_text="Residual Error (Z)", row=1, col=2)

fig.show()
```

The optimal estimator uses the shrinkage factor

This means only \(20\%\) of the observed power reading is used as the estimated temperature.

The residual points are spread around zero without a clear upward or downward pattern. This supports the orthogonality result:

\mathrm{Cov}(Z,Y)=0.

This means the remaining error is uncorrelated with the observed power reading \(Y\). Therefore, the estimator has used all useful information available from \(Y\).

The simulation also shows why \(X\) is not \(\sigma(Y)\)-measurable. For one fixed value of \(Y\), there can still be many possible values of \(X\). This is because different combinations of \(X\) and \(Q\) can produce the same value of \(Y\).
