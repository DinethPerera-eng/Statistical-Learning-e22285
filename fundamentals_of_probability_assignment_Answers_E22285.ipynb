# Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.

1. Model the experiment assuming the point is uniformly distributed with respect to area, and compute $P(A)$.
2. Model the experiment assuming the distance to the point from the center is chosen uniformly from $[0,10]$, and the direction is chosen independently and uniformly from $[0,2\pi)$ and compute $P(A)$.
3. Explain why the two answers differ.


### Answer for Q1

Let the disk of radius $10$ be represented by

$$
\Omega=\{(x,y)\in \mathbb{R}^2 : x^2+y^2\leq 10^2\}.
$$

The event $A$ means that the chosen point lies within $1$ unit of the boundary of the disk. Since the boundary is the circle with radius $10$, the point must be at a distance between $9$ and $10$ from the center. Therefore,

$$
A=\left\{(x,y)\in \Omega : 9\leq \sqrt{x^2+y^2}\leq 10\right\}.
$$

So, $A$ is the outer annulus of the disk.

---

## 1. Uniform distribution with respect to area

In this model, the point is chosen uniformly over the area of the disk. This means that regions with equal area have equal probability.

The total area of the disk is

$$
\pi(10)^2=100\pi.
$$

The event $A$ is the annulus between radius $9$ and radius $10$. Its area is

$$
\pi(10)^2-\pi(9)^2
=100\pi-81\pi
=19\pi.
$$

Therefore,

$$
P(A)=\frac{\text{area of }A}{\text{area of disk}}
=\frac{19\pi}{100\pi}
=\frac{19}{100}.
$$

Hence, under the area-uniform model,

$$
\boxed{P(A)=\frac{19}{100}}.
$$

---

## 2. Uniform distribution of the radius

In this model, the point is generated using two values:

1. The radius $r$, chosen uniformly from $[0,10]$.
2. The angle $\theta$, chosen independently and uniformly from $[0,2\pi)$.

The angle only decides the direction of the point. Since event $A$ depends on the distance from the center, we only need to consider the radius.

For event $A$, the radius must satisfy

$$
9\leq r\leq 10.
$$

The full possible interval for the radius is $[0,10]$, which has length

$$
10-0=10.
$$

The favourable interval for event $A$ is $[9,10]$, which has length

$$
10-9=1.
$$

Since the radius is chosen uniformly,

$$
P(A)=\frac{\text{favourable length}}{\text{total length}}
=\frac{1}{10}.
$$

Hence, under the radius-uniform model,

$$
\boxed{P(A)=\frac{1}{10}}.
$$

---

## 3. Why the two answers differ

The two probabilities are different because the phrase “chosen at random from a disk” can be interpreted using different probability models.

In the first model, probability is distributed uniformly over area. Therefore, the probability depends on the area of the outer annulus. This gives

$$
P(A)=\frac{19}{100}.
$$

In the second model, probability is distributed uniformly over the radius. Therefore, each interval of radius with the same length has the same probability, even though the corresponding areas are not equal. Since the interval $[9,10]$ has length $1$ out of the total radius length $10$, this gives

$$
P(A)=\frac{1}{10}.
$$

Therefore, the event $A$ is the same in both models, but the probability is different because the method of assigning probability is different.

# Q2) Bertrand's Paradox

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

Show that different mathematically natural models of “random chord” lead to different answers.

### Answer for Q2

Consider the circle

$$
x^2+y^2=1.
$$

For this unit circle, the side length of the inscribed equilateral triangle is

$$
\sqrt{3}.
$$

Before considering different random chord methods, we first identify when a chord is longer than $\sqrt{3}$.

Suppose a chord is at perpendicular distance $r$ from the center of the circle, where $0\leq r\leq 1$. Then half of the chord, the distance $r$, and the radius of the circle form a right triangle.

Since the radius is $1$, half of the chord length is

$$
\sqrt{1-r^2}.
$$

Therefore, the full chord length is

$$
L(r)=2\sqrt{1-r^2}.
$$

We need the chord length to be greater than the side length of the inscribed equilateral triangle:

$$
L(r)>\sqrt{3}.
$$

Substituting $L(r)$,

$$
2\sqrt{1-r^2}>\sqrt{3}.
$$

Squaring both sides,

$$
4(1-r^2)>3.
$$

So,

$$
4-4r^2>3
$$

$$
1>4r^2
$$

$$
r^2<\frac{1}{4}.
$$

Since $r\geq 0$,

$$
r<\frac{1}{2}.
$$

Thus, a chord is longer than $\sqrt{3}$ exactly when its perpendicular distance from the center is less than $\frac{1}{2}$.

So the required event is

$$
A=\{\text{distance of chord from center}<\frac{1}{2}\}.
$$

---

## Model 1: Choosing two random endpoints on the circumference

In this model, a chord is formed by choosing two points independently and uniformly on the circumference of the circle.

Fix one endpoint of the chord. The chord will be longer than $\sqrt{3}$ only if the second endpoint lies on the arc opposite the first point that subtends an angle of $120^\circ$.

The full circle has angle

$$
360^\circ.
$$

The favorable arc has angle

$$
120^\circ.
$$

Therefore,

$$
P(A)=\frac{120^\circ}{360^\circ}
=\frac{1}{3}.
$$

Hence, under the random-endpoints model,

$$
\boxed{P(A)=\frac{1}{3}}.
$$

---

## Model 2: Choosing the perpendicular distance uniformly

In this model, the orientation of the chord is chosen randomly, and the perpendicular distance $r$ from the center is chosen uniformly from $[0,1]$.

From the geometric condition above, the chord is longer than $\sqrt{3}$ when

$$
r<\frac{1}{2}.
$$

Since $r$ is uniformly distributed on $[0,1]$, the probability is

$$
P(A)=\frac{\frac{1}{2}-0}{1-0}
=\frac{1}{2}.
$$

Hence, under the uniform-distance model,

$$
\boxed{P(A)=\frac{1}{2}}.
$$

---

## Model 3: Choosing the midpoint of the chord uniformly inside the circle

In this model, the midpoint of the chord is chosen uniformly from the area of the circle.

The distance from the center to the chord is the same as the distance from the center to the midpoint of the chord. Therefore, the chord is longer than $\sqrt{3}$ when the midpoint lies inside the smaller circle of radius $\frac{1}{2}$.

The area of the smaller circle is

$$
\pi\left(\frac{1}{2}\right)^2
=\frac{\pi}{4}.
$$

The area of the whole unit circle is

$$
\pi(1)^2=\pi.
$$

Therefore,

$$
P(A)=\frac{\frac{\pi}{4}}{\pi}
=\frac{1}{4}.
$$

Hence, under the random-midpoint model,

$$
\boxed{P(A)=\frac{1}{4}}.
$$

---

## Conclusion

The probabilities are different because the phrase “a chord is chosen at random” does not define one unique probability model.

The three natural methods give

$$
\boxed{\frac{1}{3}},\qquad
\boxed{\frac{1}{2}},\qquad
\boxed{\frac{1}{4}}.
$$

This shows Bertrand's paradox. The same geometric question gives different answers depending on how the random chord is generated.

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?

### Answer

The sample space is

$$
\Omega=\{BB,BG,GB,GG\},
$$

where the first letter represents the older child and the second letter represents the younger child.

Each outcome has probability

$$
\frac{1}{4}.
$$

Given that at least one child is a girl, the outcome $BB$ is excluded. Therefore, the remaining outcomes are

$$
\{BG,GB,GG\}.
$$

Among these three outcomes, only $GG$ represents both children being girls.

Therefore,

$$
P(\text{both girls}\mid \text{at least one girl})
=
\frac{1}{3}.
$$

Hence,

$$
\boxed{\frac{1}{3}}.
$$

# Q9) Information Available in a Discrete Random Variable

A factory inspects a product and records its quality using two different coding systems.

Each product falls into exactly one of the following four categories:

* $G_1$: good quality, produced on Machine 1
* $G_2$: good quality, produced on Machine 2
* $D_1$: defective, produced on Machine 1
* $D_2$: defective, produced on Machine 2

Assume the sample space is

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

with probabilities

$$
P(\{G_1\})=0.50,\qquad P(\{G_2\})=0.20,\qquad P(\{D_1\})=0.10,\qquad P(\{D_2\})=0.20.
$$

Define two discrete random variables on the same probability space:

* $X$ records only whether the item is good or defective:
  $$
  X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.
  $$

* $Y$ records both quality and machine:
  $$
  Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4.
  $$

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that $\sigma(X)\subseteq \sigma(Y)$, and explain why $Y$ contains more information than $X$.

5. Find the marginal distribution of $X$.

6. Find the marginal distribution of $Y$.

7. Compute the joint probabilities
   $$
   P(X=0,Y=1),\quad P(X=0,Y=2),\quad P(X=1,Y=3),\quad P(X=1,Y=4).
   $$

8. Compute the conditional probabilities
   $$
   P(X=1\mid Y=3),\qquad P(X=1\mid Y=4),\qquad P(Y=3\mid X=1),\qquad P(Y=4\mid X=1).
   $$

9. Interpret the difference between conditioning on $X$ and conditioning on $Y$.

10. Explain, in words, what information is available in $\sigma(X)$ but not in $\sigma(Y)$, and what information is available in $\sigma(Y)$ but not in $\sigma(X)$.

11. Give a Python based visualization of the model.

12. Find the marginal, joint, and conditional Shannon Entropy of $X$ and $Y$.

### Answer

#### 1. Probability space

The probability space is given by

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

$$
\mathfrak F=\mathcal P(\Omega),
$$

with probability measure $P$ defined as

$$
P(\{G_1\})=0.50,\qquad
P(\{G_2\})=0.20,\qquad
P(\{D_1\})=0.10,\qquad
P(\{D_2\})=0.20.
$$

Also,

$$
0.50+0.20+0.10+0.20=1.
$$

Hence this is a valid probability space.

---

#### 2. Sigma-algebra generated by $X$

The random variable $X$ records only the quality type:

$$
X(G_1)=0,\qquad X(G_2)=0,\qquad X(D_1)=1,\qquad X(D_2)=1.
$$

So the inverse images are

$$
X^{-1}(\{0\})=\{G_1,G_2\},
$$

and

$$
X^{-1}(\{1\})=\{D_1,D_2\}.
$$

Therefore,

$$
\sigma(X)=\left\{\varnothing,\Omega,\{G_1,G_2\},\{D_1,D_2\}\right\}.
$$

---

#### 3. Sigma-algebra generated by $Y$

The random variable $Y$ records both the quality and the machine:

$$
Y(G_1)=1,\qquad Y(G_2)=2,\qquad Y(D_1)=3,\qquad Y(D_2)=4.
$$

Each outcome in $\Omega$ is mapped to a different value, so $Y$ separates all points of the sample space. Hence

$$
\sigma(Y)=\mathcal P(\Omega).
$$

Explicitly, this is the full power set of $\Omega$.

---

#### 4. Show that $\sigma(X)\subseteq \sigma(Y)$

Since

$$
\sigma(Y)=\mathcal P(\Omega),
$$

every event in $\sigma(X)$ is automatically contained in $\sigma(Y)$. Therefore,

$$
\sigma(X)\subseteq \sigma(Y).
$$

The variable $X$ only tells whether the item is good or defective, while $Y$ tells both the quality and the machine number. So $Y$ contains more information than $X$.

---

#### 5. Marginal distribution of $X$

For $X=0$,

$$
P(X=0)=P(\{G_1,G_2\})=0.50+0.20=0.70.
$$

For $X=1$,

$$
P(X=1)=P(\{D_1,D_2\})=0.10+0.20=0.30.
$$

So the marginal distribution of $X$ is

$$
P(X=0)=0.70,\qquad P(X=1)=0.30.
$$

---

#### 6. Marginal distribution of $Y$

The marginal probabilities for $Y$ are

$$
P(Y=1)=P(\{G_1\})=0.50,
$$

$$
P(Y=2)=P(\{G_2\})=0.20,
$$

$$
P(Y=3)=P(\{D_1\})=0.10,
$$

$$
P(Y=4)=P(\{D_2\})=0.20.
$$

---

#### 7. Joint probabilities

The required joint probabilities are

$$
P(X=0,Y=1)=P(\{G_1\})=0.50,
$$

$$
P(X=0,Y=2)=P(\{G_2\})=0.20,
$$

$$
P(X=1,Y=3)=P(\{D_1\})=0.10,
$$

$$
P(X=1,Y=4)=P(\{D_2\})=0.20.
$$

All remaining joint probabilities are zero.

---

#### 8. Conditional probabilities

Since $Y=3$ means the outcome is $D_1$, we get

$$
P(X=1\mid Y=3)=1.
$$

Similarly, $Y=4$ means the outcome is $D_2$, so

$$
P(X=1\mid Y=4)=1.
$$

Next,

$$
P(Y=3\mid X=1)
=
\frac{P(Y=3,X=1)}{P(X=1)}
=
\frac{0.10}{0.30}
=
\frac{1}{3}.
$$

Also,

$$
P(Y=4\mid X=1)
=
\frac{P(Y=4,X=1)}{P(X=1)}
=
\frac{0.20}{0.30}
=
\frac{2}{3}.
$$

---

#### 9. Interpretation of conditioning on $X$ and on $Y$

Conditioning on $Y$ gives more detailed information than conditioning on $X$.

For example, if $Y=3$, then the exact outcome is known and the item must be defective. So

$$
P(X=1\mid Y=3)=1.
$$

But if we only know that $X=1$, we know the item is defective but we still do not know whether it came from Machine 1 or Machine 2. Therefore,

$$
P(Y=3\mid X=1)=\frac{1}{3},\qquad
P(Y=4\mid X=1)=\frac{2}{3}.
$$

---

#### 10. Information in $\sigma(X)$ and $\sigma(Y)$

The sigma-algebra $\sigma(X)$ contains only information about whether the item is good or defective. It does not distinguish between Machine 1 and Machine 2.

The sigma-algebra $\sigma(Y)$ contains complete information about the outcome, since it tells both the quality and the machine. Therefore, $\sigma(Y)$ is finer than $\sigma(X)$.

$$
H(Z)=-\sum_z P(z)\ln P(z).

Therefore,

$$
H(X)\approx 0.61086.
$$

$$

Hence,

$$
H(Y)\approx 1.22061.
$$

For the joint entropy,

$$
H(X,Y)=-\sum_{x,y}P(x,y)\ln P(x,y).
$$

Only the nonzero joint probabilities contribute, so

$$
H(X,Y)=-(0.50\ln 0.50+0.20\ln 0.20+0.10\ln 0.10+0.20\ln 0.20).
$$

Thus,

$$
H(X,Y)\approx 1.22061.

Since each value of $Y$ determines $X$ exactly,

$$
H(X\mid Y)=0.

Also,

$$
H(Y\mid X)=H(X,Y)-H(X).

So,

$$
H(Y\mid X)\approx 1.22061-0.61086=0.60975.

# Q10) Information Available in Continuous Random Varaibles

A chemical plant monitors each production run using two different measurement systems.

For a given run, let

* $T$ be the deviation of the reactor temperature from target, measured in standardized units,
* $P$ be the deviation of the reactor pressure from target, also measured in standardized units.

Assume that $(T,P)$ is jointly continuous, with independent normal components:

$$
T \sim N(0,1), \qquad P \sim N(0,4),
$$

and $T$ and $P$ are independent.

Let the underlying probability space be

$$
\Omega=\mathbb R^2, \qquad \mathfrak F=\mathfrak B(\mathbb R^2),
$$

and let $P$ be the probability measure induced by the joint density

$$
f_{T,P}(t,p)=\frac{1}{4\pi}\exp\left(-\frac{t^2}{2}-\frac{p^2}{8}\right), \qquad (t,p)\in\mathbb R^2.
$$

Define two random variables on the same probability space:

* $X:\Omega\to\mathbb R$ by
  $$
  X(t,p)=t,
  $$
  so that $X$ records only the reactor temperature deviation;

* $Y:\Omega\to\mathbb R^2$ by
  $$
  Y(t,p)=(t,p),
  $$
  so that $Y$ records the full sensor state: both temperature and pressure deviations.

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$ explicitly.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that
   $$
   \sigma(X)\subseteq \sigma(Y),
   $$
   and explain why $Y$ contains more information than $X$.

5. Find the marginal density of $X$.

6. Find the marginal density of $Y$.

7. Compute the following probabilities:
   $$
   P(X\leq 0), \qquad P(X>1), \qquad P(Y\in (-\infty,0]\times\mathbb R), \qquad P(Y\in [ -1,1]\times[-2,2]).
   $$

8. Compute the conditional distribution of $X$ given $Y=(t,p)$.

9. Compute the conditional distribution of $Y$ given $X=t$.

10. Explain, in words, what is known when one conditions on $X$ and what additional information becomes available when one conditions on $Y$.

11. Show that
    $$
    X=\pi_1\circ Y,
    $$
    where $\pi_1(t,p)=t$, and explain why this implies that $X$ is a measurable function of $Y$.

12. Give a Python-based visualization of the model, for example by plotting:

* the joint density of $Y=(T,P)$ as a contour plot or heat map;
* the marginal density of $X$;
* and a diagram illustrating that $\sigma(X)$ is coarser than $\sigma(Y)$.

13. Compute the differential entropy of $X$:

$$
h(X)=-\int_{\mathbb R} f_X(x)\ln f_X(x)\,dx.
$$

14. Compute the differential entropy of $Y$:

$$
h(Y)=-\int_{\mathbb R^2} f_Y(t,p)\ln f_Y(t,p)\,dt\,dp.
$$

15. Compute the conditional differential entropy

$$
h(Y|X),
$$

and interpret it physically.

16. Discuss what happens to

$$
h(X|Y)
$$

in this model, and explain why the continuous case differs from the discrete case when one random variable is completely determined by the other.


### Answer

#### 1. Probability space

The sample space and event space are

$$
\Omega=\mathbb R^2,\qquad \mathfrak F=\mathfrak B(\mathbb R^2).
$$

Here, $\mathfrak B(\mathbb R^2)$ denotes the Borel $\sigma$-algebra on $\mathbb R^2$.

A single outcome can be written as

$$
\omega=(t,p),
$$

where $t$ is the standardized temperature deviation and $p$ is the standardized pressure deviation.

The probability measure is determined by the joint density

$$
f_{T,P}(t,p)=\frac{1}{4\pi}\exp\left(-\frac{t^2}{2}-\frac{p^2}{8}\right),
\qquad (t,p)\in\mathbb R^2.
$$

Therefore, for any Borel set $A\subseteq\mathbb R^2$,

$$
P(A)=\iint_A f_{T,P}(t,p)\,dt\,dp.
$$

Since $T\sim N(0,1)$ and $P\sim N(0,4)$ are independent, this gives a valid probability measure.

---

#### 2. Sigma-algebra generated by $X$

The random variable $X$ is

$$
X(t,p)=t.
$$

So $X$ observes only the temperature coordinate.

The $\sigma$-algebra generated by $X$ is

$$
\sigma(X)=\{X^{-1}(B):B\in\mathfrak B(\mathbb R)\}.
$$

For any Borel set $B\subseteq\mathbb R$,

$$
X^{-1}(B)=\{(t,p)\in\mathbb R^2:t\in B\}=B\times\mathbb R.
$$

Hence,

$$
\sigma(X)=\{B\times\mathbb R:B\in\mathfrak B(\mathbb R)\}.
$$

Thus, $\sigma(X)$ contains events that depend only on the temperature value.

---

#### 3. Sigma-algebra generated by $Y$

The random variable $Y$ is

$$
Y(t,p)=(t,p).
$$

This means $Y$ records both temperature and pressure.

Since $Y$ is the identity map on $\mathbb R^2$, for any Borel set $A\subseteq\mathbb R^2$,

$$
Y^{-1}(A)=A.
$$

Therefore,

$$
\sigma(Y)=\mathfrak B(\mathbb R^2).
$$

So $Y$ generates the full Borel $\sigma$-algebra on the plane.

---

#### 4. Showing $\sigma(X)\subseteq\sigma(Y)$

Every set of the form $B\times\mathbb R$ is a Borel subset of $\mathbb R^2$. Therefore,

$$
\sigma(X)\subseteq\sigma(Y).
$$

This inclusion is strict because a set such as

$$
[-1,1]\times[-2,2]
$$

belongs to $\sigma(Y)$, but it is not in $\sigma(X)$ because it depends on both $t$ and $p$.

Therefore, $X$ gives only the temperature deviation, while $Y$ gives both the temperature and pressure deviations.

---

#### 5. Marginal density of $X$

The marginal density of $X$ is found by integrating out the pressure coordinate:

$$
f_X(x)=\int_{-\infty}^{\infty}f_{T,P}(x,p)\,dp.
$$

Substituting the joint density,

$$
f_X(x)=\int_{-\infty}^{\infty}
\frac{1}{4\pi}
\exp\left(-\frac{x^2}{2}-\frac{p^2}{8}\right)\,dp.
$$

Using the normal density of $P\sim N(0,4)$, this simplifies to

$$
f_X(x)=\frac{1}{\sqrt{2\pi}}\exp\left(-\frac{x^2}{2}\right).
$$

Therefore,

$$
X\sim N(0,1).
$$

---

#### 6. Marginal density of $Y$

Since $Y=(T,P)$, the density of $Y$ is the joint density of $T$ and $P$:

$$
f_Y(t,p)=\frac{1}{4\pi}\exp\left(-\frac{t^2}{2}-\frac{p^2}{8}\right).
$$

Thus, $Y$ is a bivariate normal random vector with independent components.

---

#### 7. Required probabilities

By symmetry of the standard normal distribution,

$$
P(X\leq 0)=P(T\leq 0)=\frac{1}{2}.
$$

Also,

$$
P(X>1)=P(T>1)=1-\Phi(1),
$$

where $\Phi$ is the standard normal distribution function. Numerically,

$$
P(X>1)\approx 0.1587.
$$

Next,

$$
P\bigl(Y\in(-\infty,0]\times\mathbb R\bigr)=P(T\leq 0)=\frac{1}{2}.
$$

For the rectangle,

$$
P\bigl(Y\in[-1,1]\times[-2,2]\bigr)
=
P(-1\leq T\leq 1,\,-2\leq P\leq 2).
$$

Since $T$ and $P$ are independent,

$$
P\bigl(Y\in[-1,1]\times[-2,2]\bigr)
=
P(-1\leq T\leq 1)P(-2\leq P\leq 2).
$$

Now,

$$
P(-1\leq T\leq 1)=2\Phi(1)-1.
$$

Since $P\sim N(0,4)$, we have $P/2\sim N(0,1)$. Therefore,

$$
P(-2\leq P\leq 2)=P(-1\leq P/2\leq 1)=2\Phi(1)-1.
$$

Hence,

$$
P\bigl(Y\in[-1,1]\times[-2,2]\bigr)
=
\bigl(2\Phi(1)-1\bigr)^2.
$$

Numerically,

$$
2\Phi(1)-1\approx 0.6827,
$$

so

$$
P\bigl(Y\in[-1,1]\times[-2,2]\bigr)\approx 0.466.
$$

---

#### 8. Conditional distribution of $X$ given $Y=(t,p)$

Since $Y=(T,P)$ gives the full pair of values, it also gives the value of $T$ exactly.

As $X=T$, the conditional distribution of $X$ given $Y=(t,p)$ is concentrated at $t$:

$$
P(X=t\mid Y=(t,p))=1.
$$

Equivalently,

$$
X\mid Y=(t,p)\sim \delta_t,
$$

where $\delta_t$ is the Dirac mass at $t$.

---

#### 9. Conditional distribution of $Y$ given $X=t$

Given $X=t$, the temperature coordinate is fixed at $t$.

Since $T$ and $P$ are independent, the pressure coordinate remains normally distributed as

$$
P\sim N(0,4).
$$

Therefore,

$$
Y\mid X=t\sim (t,P).
$$

Equivalently, the conditional probability measure is

$$
P_{Y\mid X=t}=\delta_t\otimes \mathcal N(0,4).
$$

Formally, this can be written as

$$
f_{Y\mid X}\bigl((u,v)\mid t\bigr)
=
\delta_t(u)\cdot
\frac{1}{\sqrt{8\pi}}
\exp\left(-\frac{v^2}{8}\right).
$$

So, after conditioning on $X=t$, the temperature is fixed, but the pressure remains random.

---

#### 10. Meaning of conditioning on $X$ and $Y$

Conditioning on $X$ means that the temperature deviation is known, but the pressure deviation is still unknown.

Conditioning on $Y$ means that both temperature and pressure deviations are known.

So:

* given $X=t$, the pressure is still random;
* given $Y=(t,p)$, both coordinates are known exactly.

---

#### 11. Showing $X=\pi_1\circ Y$

Define the projection map

$$
\pi_1:\mathbb R^2\to\mathbb R,
\qquad
\pi_1(t,p)=t.
$$

Then,

$$
X=\pi_1\circ Y.
$$

This is because

$$
(\pi_1\circ Y)(t,p)=\pi_1(t,p)=t=X(t,p).
$$

Since $\pi_1$ is Borel measurable, $X$ is a measurable function of $Y$. Therefore,

$$
\sigma(X)\subseteq\sigma(Y).
$$

This explains mathematically why $Y$ contains at least as much information as $X$.

---

#### 12. Python-based visualization

A Python visualization can show the following:

* the joint density of $Y=(T,P)$,

$$
f_Y(t,p)=\frac{1}{4\pi}\exp\left(-\frac{t^2}{2}-\frac{p^2}{8}\right);
$$

* the marginal density of $X$,

$$
f_X(x)=\frac{1}{\sqrt{2\pi}}e^{-x^2/2};
$$

* vertical strips of the form

$$
B\times\mathbb R,
$$

which represent events in $\sigma(X)$;

* and two-dimensional Borel regions such as

$$
[-1,1]\times[-2,2],
$$

which represent events in $\sigma(Y)$.

This illustrates that $\sigma(X)$ is coarser because it only observes vertical strips, while $\sigma(Y)$ can observe full two-dimensional events.

---

#### 13. Differential entropy of $X$

Since $X\sim N(0,1)$,

$$
h(X)=\frac{1}{2}\ln(2\pi e).
$$

Numerically,

$$
h(X)\approx 1.4189.
$$

---

#### 14. Differential entropy of $Y$

Since $Y=(T,P)$ and the components are independent normal variables,

$$
h(Y)=h(T)+h(P).
$$

For $T\sim N(0,1)$,

$$
h(T)=\frac{1}{2}\ln(2\pi e).
$$

For $P\sim N(0,4)$,

$$
h(P)=\frac{1}{2}\ln(2\pi e\cdot 4).
$$

Therefore,

$$
h(Y)=\frac{1}{2}\ln(2\pi e)+\frac{1}{2}\ln(8\pi e).
$$

Equivalently,

$$
h(Y)=\frac{1}{2}\ln\bigl((2\pi e)^2\cdot 4\bigr)
=
\ln(2\pi e)+\ln 2.
$$

Numerically,

$$
h(Y)\approx 3.5310.
$$

---

#### 15. Conditional differential entropy $h(Y\mid X)$

Since $X=T$ and $Y=(T,P)$, knowing $X$ means the temperature part is already known.

The only remaining uncertainty in $Y$ is the pressure component $P$. Therefore,

$$
h(Y\mid X)=h(P).
$$

Since $P\sim N(0,4)$,

$$
h(Y\mid X)=\frac{1}{2}\ln(2\pi e\cdot 4).
$$

So,

$$
h(Y\mid X)=\frac{1}{2}\ln(8\pi e).
$$

Numerically,

$$
h(Y\mid X)\approx 2.1121.
$$

Physically, this means that once the temperature deviation is known, the remaining uncertainty is the pressure deviation.

---

#### 16. Discussion of $h(X\mid Y)$

Since $X$ is a deterministic function of $Y$, in a discrete setting we would have

$$
H(X\mid Y)=0.
$$

However, this is a continuous model, and differential entropy behaves differently.

Here, the conditional distribution of $X$ given $Y=(t,p)$ is degenerate at $t$. This is not an ordinary continuous distribution with a density with respect to Lebesgue measure.

Therefore, the conditional differential entropy is not a finite ordinary value. It is usually understood formally as

$$
h(X\mid Y)=-\infty.
$$

So, unlike the discrete case, deterministic dependence in the continuous case does not give conditional differential entropy equal to zero.

This shows the difference between discrete entropy and differential entropy in continuous probability models.

##Python Visualization

##The  σ -algebra visualizations

For this model,

$$
X(t,p)=t,\qquad Y(t,p)=(t,p).
$$

The two generated sigma-algebras can be visualized in the plane $\mathbb R^2$.

The sigma-algebra generated by $X$ contains events of the form

$$
B\times \mathbb R,
$$

where $B\in\mathfrak B(\mathbb R)$. These events depend only on the temperature coordinate $t$, so they appear as vertical strips in the $(t,p)$ plane.

On the other hand,

$$
\sigma(Y)=\mathfrak B(\mathbb R^2),
$$

because $Y$ records both coordinates. Therefore, events in $\sigma(Y)$ can be general two-dimensional Borel regions, such as rectangles, disks, or other regions in the plane.

The following Python visualization shows:

1. a vertical strip representing an event in $\sigma(X)$;
2. a two-dimensional region representing an event in $\sigma(Y)$;
3. a comparison showing that $\sigma(X)$ is coarser than $\sigma(Y)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle

# Create figure with three panels
fig, axs = plt.subplots(1, 3, figsize=(15, 4.8))

# Common axis limits
t_min, t_max = -4, 4
p_min, p_max = -8, 8

# =====================================================
# 1. Example of an event in sigma(X)
# =====================================================
ax = axs[0]
ax.set_title(r"Event in $\sigma(X)$: $B\times\mathbb{R}$")
ax.set_xlim(t_min, t_max)
ax.set_ylim(p_min, p_max)
ax.set_xlabel(r"$t$")
ax.set_ylabel(r"$p$")

# Example strip: B = [-1,1]
event_x = Rectangle(
    (-1, p_min), 2, p_max - p_min,
    facecolor="skyblue", edgecolor="navy", alpha=0.55
)
ax.add_patch(event_x)

ax.text(
    0, 0, r"$[-1,1]\times\mathbb{R}$",
    ha="center", va="center", fontsize=12,
    bbox=dict(facecolor="white", edgecolor="none", alpha=0.8)
)

ax.axhline(0, color="black", lw=0.8)
ax.axvline(0, color="black", lw=0.8)
ax.grid(alpha=0.25)

# =====================================================
# 2. Example of an event in sigma(Y)
# =====================================================
ax = axs[1]
ax.set_title(r"Event in $\sigma(Y)$: general Borel subset")
ax.set_xlim(t_min, t_max)
ax.set_ylim(p_min, p_max)
ax.set_xlabel(r"$t$")
ax.set_ylabel(r"$p$")

# Example rectangle
event_y_rect = Rectangle(
    (-1.5, -3), 3, 6,
    facecolor="lightgreen", edgecolor="darkgreen", alpha=0.55
)
ax.add_patch(event_y_rect)

# Example disk
event_y_disk = Circle(
    (2.2, 3.5), 1.2,
    facecolor="salmon", edgecolor="darkred", alpha=0.45
)
ax.add_patch(event_y_disk)

ax.text(
    0, 0, r"$[-1.5,1.5]\times[-3,3]$",
    ha="center", va="center", fontsize=11,
    bbox=dict(facecolor="white", edgecolor="none", alpha=0.8)
)
ax.text(
    2.2, 3.5, "disk",
    ha="center", va="center", fontsize=10,
    bbox=dict(facecolor="white", edgecolor="none", alpha=0.8)
)

ax.axhline(0, color="black", lw=0.8)
ax.axvline(0, color="black", lw=0.8)
ax.grid(alpha=0.25)

# =====================================================
# 3. Comparison of sigma(X) and sigma(Y)
# =====================================================
ax = axs[2]
ax.set_title(r"Comparison: $\sigma(X)\subseteq\sigma(Y)$")
ax.set_xlim(t_min, t_max)
ax.set_ylim(p_min, p_max)
ax.set_xlabel(r"$t$")
ax.set_ylabel(r"$p$")

# Event from sigma(X)
compare_strip = Rectangle(
    (-1, p_min), 2, p_max - p_min,
    facecolor="skyblue", edgecolor="navy", alpha=0.35,
    label=r"$\sigma(X)$ event"
)
ax.add_patch(compare_strip)

# Event from sigma(Y)
compare_rect = Rectangle(
    (-1.5, -3), 3, 6,
    facecolor="lightgreen", edgecolor="darkgreen", alpha=0.50,
    label=r"$\sigma(Y)$ event"
)
ax.add_patch(compare_rect)

ax.text(
    0, 6.5, r"$\sigma(X)$ sees only vertical strips",
    ha="center", fontsize=11,
    bbox=dict(facecolor="white", edgecolor="none", alpha=0.8)
)
ax.text(
    0, -6.5, r"$\sigma(Y)$ sees full 2D regions",
    ha="center", fontsize=11,
    bbox=dict(facecolor="white", edgecolor="none", alpha=0.8)
)

ax.axhline(0, color="black", lw=0.8)
ax.axvline(0, color="black", lw=0.8)
ax.legend(loc="upper right")
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

The distributions

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# =====================================================
# Parameters
# =====================================================
mu_T, sigma_T = 0.0, 1.0
mu_P, sigma_P = 0.0, 2.0   # because Var(P)=4

# Grid values
t_vals = np.linspace(-4, 4, 400)
p_vals = np.linspace(-8, 8, 400)
T_grid, P_grid = np.meshgrid(t_vals, p_vals)

# =====================================================
# Densities
# =====================================================

# Joint density of Y=(T,P)
joint_density = (
    norm.pdf(T_grid, loc=mu_T, scale=sigma_T)
    *
    norm.pdf(P_grid, loc=mu_P, scale=sigma_P)
)

# Marginal density of X=T
f_X = norm.pdf(t_vals, loc=mu_T, scale=sigma_T)

# Marginal density of P
f_P = norm.pdf(p_vals, loc=mu_P, scale=sigma_P)

# Because T and P are independent, P | X=t0 has the same distribution as P
t0 = 1.0
f_P_given_X = norm.pdf(p_vals, loc=mu_P, scale=sigma_P)

# X | Y=(t1,p1) is degenerate at t1
# For plotting only, use a very narrow Gaussian approximation
t1, p1 = 1.0, 2.0
eps = 0.08
f_X_given_Y_vis = norm.pdf(t_vals, loc=t1, scale=eps)

# =====================================================
# Figure layout
# =====================================================
fig = plt.figure(figsize=(12, 12))
grid = fig.add_gridspec(
    3, 2,
    height_ratios=[1.1, 1, 1],
    hspace=0.45,
    wspace=0.28
)

# =====================================================
# 1. Joint density of Y=(T,P)
# =====================================================
ax1 = fig.add_subplot(grid[0, :])
filled = ax1.contourf(T_grid, P_grid, joint_density, levels=25, cmap="viridis")
ax1.contour(T_grid, P_grid, joint_density, levels=10, colors="white", linewidths=0.5, alpha=0.6)
ax1.set_title(r"(1) Joint density of $Y=(T,P)$")
ax1.set_xlabel(r"$t$")
ax1.set_ylabel(r"$p$")
fig.colorbar(filled, ax=ax1, shrink=0.9, label="Density")

# =====================================================
# 2. Marginal density of X
# =====================================================
ax2 = fig.add_subplot(grid[1, 0])
ax2.plot(t_vals, f_X, color="navy", lw=2)
ax2.fill_between(t_vals, f_X, color="skyblue", alpha=0.35)
ax2.set_title(r"(2) Marginal density of $X=T$")
ax2.set_xlabel(r"$x$")
ax2.set_ylabel(r"$f_X(x)$")
ax2.grid(alpha=0.25)

# =====================================================
# 3. Marginal density of P
# =====================================================
ax3 = fig.add_subplot(grid[1, 1])
ax3.plot(p_vals, f_P, color="darkgreen", lw=2)
ax3.fill_between(p_vals, f_P, color="lightgreen", alpha=0.35)
ax3.set_title(r"(3) Marginal density of $P$")
ax3.set_xlabel(r"$p$")
ax3.set_ylabel(r"$f_P(p)$")
ax3.grid(alpha=0.25)

# =====================================================
# 4. Conditional density of P | X=t0
# =====================================================
ax4 = fig.add_subplot(grid[2, 0])
ax4.plot(
    p_vals, f_P_given_X,
    color="crimson", lw=2,
    label=rf"$f_{{P\mid X={t0}}}(p)$"
)
ax4.fill_between(p_vals, f_P_given_X, color="salmon", alpha=0.35)
ax4.set_title(r"(4) Conditional density of $P\mid X=t_0$")
ax4.set_xlabel(r"$p$")
ax4.set_ylabel(r"$f_{P\mid X=t_0}(p)$")
ax4.legend()
ax4.grid(alpha=0.25)

# =====================================================
# 5. Visualization of X | Y=(t1,p1)
# =====================================================
ax5 = fig.add_subplot(grid[2, 1])
ax5.plot(
    t_vals, f_X_given_Y_vis,
    color="purple", lw=2,
    label=rf"visualization of $X\mid Y=({t1},{p1})$"
)
ax5.axvline(t1, color="black", ls="--", lw=1.3, label=rf"$x={t1}$")
ax5.set_title(r"(5) $X\mid Y=(t_1,p_1)$ is degenerate at $x=t_1$")
ax5.set_xlabel(r"$x$")
ax5.set_ylabel("Approx. density")
ax5.legend()
ax5.grid(alpha=0.25)

plt.show()
```
