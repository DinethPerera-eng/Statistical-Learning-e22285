# Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.

1. Model the experiment assuming the point is uniformly distributed with respect to area, and compute $P(A)$.
2. Model the experiment assuming the distance to the point from the center is chosen uniformly from $[0,10]$, and the direction is chosen independently and uniformly from $[0,2\pi)$ and compute $P(A)$.
3. Explain why the two answers differ.


### Answer for Q1

Let the disk of radius $10$ be represented by

$$
\Omega=\{(x,y)\in \mathbb{R}^2 : x^2+y^2\leq 10^2\}.
$$

The event $A$ means that the chosen point lies within $1$ unit of the boundary of the disk. Since the boundary is the circle with radius $10$, the point must be at a distance between $9$ and $10$ from the center. Therefore,

$$
A=\left\{(x,y)\in \Omega : 9\leq \sqrt{x^2+y^2}\leq 10\right\}.
$$

So, $A$ is the outer annulus of the disk.

---

## 1. Uniform distribution with respect to area

In this model, the point is chosen uniformly over the area of the disk. This means that regions with equal area have equal probability.

The total area of the disk is

$$
\pi(10)^2=100\pi.
$$

The event $A$ is the annulus between radius $9$ and radius $10$. Its area is

$$
\pi(10)^2-\pi(9)^2
=100\pi-81\pi
=19\pi.
$$

Therefore,

$$
P(A)=\frac{\text{area of }A}{\text{area of disk}}
=\frac{19\pi}{100\pi}
=\frac{19}{100}.
$$

Hence, under the area-uniform model,

$$
\boxed{P(A)=\frac{19}{100}}.
$$

---

## 2. Uniform distribution of the radius

In this model, the point is generated using two values:

1. The radius $r$, chosen uniformly from $[0,10]$.
2. The angle $\theta$, chosen independently and uniformly from $[0,2\pi)$.

The angle only decides the direction of the point. Since event $A$ depends on the distance from the center, we only need to consider the radius.

For event $A$, the radius must satisfy

$$
9\leq r\leq 10.
$$

The full possible interval for the radius is $[0,10]$, which has length

$$
10-0=10.
$$

The favourable interval for event $A$ is $[9,10]$, which has length

$$
10-9=1.
$$

Since the radius is chosen uniformly,

$$
P(A)=\frac{\text{favourable length}}{\text{total length}}
=\frac{1}{10}.
$$

Hence, under the radius-uniform model,

$$
\boxed{P(A)=\frac{1}{10}}.
$$

---

## 3. Why the two answers differ

The two probabilities are different because the phrase “chosen at random from a disk” can be interpreted using different probability models.

In the first model, probability is distributed uniformly over area. Therefore, the probability depends on the area of the outer annulus. This gives

$$
P(A)=\frac{19}{100}.
$$

In the second model, probability is distributed uniformly over the radius. Therefore, each interval of radius with the same length has the same probability, even though the corresponding areas are not equal. Since the interval $[9,10]$ has length $1$ out of the total radius length $10$, this gives

$$
P(A)=\frac{1}{10}.
$$

Therefore, the event $A$ is the same in both models, but the probability is different because the method of assigning probability is different.

# Q2) Bertrand's Paradox

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

Show that different mathematically natural models of “random chord” lead to different answers.

### Answer for Q2

Consider the circle

$$
x^2+y^2=1.
$$

For this unit circle, the side length of the inscribed equilateral triangle is

$$
\sqrt{3}.
$$

Before considering different random chord methods, we first identify when a chord is longer than $\sqrt{3}$.

Suppose a chord is at perpendicular distance $r$ from the center of the circle, where $0\leq r\leq 1$. Then half of the chord, the distance $r$, and the radius of the circle form a right triangle.

Since the radius is $1$, half of the chord length is

$$
\sqrt{1-r^2}.
$$

Therefore, the full chord length is

$$
L(r)=2\sqrt{1-r^2}.
$$

We need the chord length to be greater than the side length of the inscribed equilateral triangle:

$$
L(r)>\sqrt{3}.
$$

Substituting $L(r)$,

$$
2\sqrt{1-r^2}>\sqrt{3}.
$$

Squaring both sides,

$$
4(1-r^2)>3.
$$

So,

$$
4-4r^2>3
$$

$$
1>4r^2
$$

$$
r^2<\frac{1}{4}.
$$

Since $r\geq 0$,

$$
r<\frac{1}{2}.
$$

Thus, a chord is longer than $\sqrt{3}$ exactly when its perpendicular distance from the center is less than $\frac{1}{2}$.

So the required event is

$$
A=\{\text{distance of chord from center}<\frac{1}{2}\}.
$$

---

## Model 1: Choosing two random endpoints on the circumference

In this model, a chord is formed by choosing two points independently and uniformly on the circumference of the circle.

Fix one endpoint of the chord. The chord will be longer than $\sqrt{3}$ only if the second endpoint lies on the arc opposite the first point that subtends an angle of $120^\circ$.

The full circle has angle

$$
360^\circ.
$$

The favorable arc has angle

$$
120^\circ.
$$

Therefore,

$$
P(A)=\frac{120^\circ}{360^\circ}
=\frac{1}{3}.
$$

Hence, under the random-endpoints model,

$$
\boxed{P(A)=\frac{1}{3}}.
$$

---

## Model 2: Choosing the perpendicular distance uniformly

In this model, the orientation of the chord is chosen randomly, and the perpendicular distance $r$ from the center is chosen uniformly from $[0,1]$.

From the geometric condition above, the chord is longer than $\sqrt{3}$ when

$$
r<\frac{1}{2}.
$$

Since $r$ is uniformly distributed on $[0,1]$, the probability is

$$
P(A)=\frac{\frac{1}{2}-0}{1-0}
=\frac{1}{2}.
$$

Hence, under the uniform-distance model,

$$
\boxed{P(A)=\frac{1}{2}}.
$$

---

## Model 3: Choosing the midpoint of the chord uniformly inside the circle

In this model, the midpoint of the chord is chosen uniformly from the area of the circle.

The distance from the center to the chord is the same as the distance from the center to the midpoint of the chord. Therefore, the chord is longer than $\sqrt{3}$ when the midpoint lies inside the smaller circle of radius $\frac{1}{2}$.

The area of the smaller circle is

$$
\pi\left(\frac{1}{2}\right)^2
=\frac{\pi}{4}.
$$

The area of the whole unit circle is

$$
\pi(1)^2=\pi.
$$

Therefore,

$$
P(A)=\frac{\frac{\pi}{4}}{\pi}
=\frac{1}{4}.
$$

Hence, under the random-midpoint model,

$$
\boxed{P(A)=\frac{1}{4}}.
$$

---

## Conclusion

The probabilities are different because the phrase “a chord is chosen at random” does not define one unique probability model.

The three natural methods give

$$
\boxed{\frac{1}{3}},\qquad
\boxed{\frac{1}{2}},\qquad
\boxed{\frac{1}{4}}.
$$

This shows Bertrand's paradox. The same geometric question gives different answers depending on how the random chord is generated.

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?

### Answer

The sample space is

$$
\Omega=\{BB,BG,GB,GG\},
$$

where the first letter represents the older child and the second letter represents the younger child.

Each outcome has probability

$$
\frac{1}{4}.
$$

Given that at least one child is a girl, the outcome $BB$ is excluded. Therefore, the remaining outcomes are

$$
\{BG,GB,GG\}.
$$

Among these three outcomes, only $GG$ represents both children being girls.

Therefore,

$$
P(\text{both girls}\mid \text{at least one girl})
=
\frac{1}{3}.
$$

Hence,

$$
\boxed{\frac{1}{3}}.
$$

# Q9) Information Available in a Discrete Random Variable

A factory inspects a product and records its quality using two different coding systems.

Each product falls into exactly one of the following four categories:

* $G_1$: good quality, produced on Machine 1
* $G_2$: good quality, produced on Machine 2
* $D_1$: defective, produced on Machine 1
* $D_2$: defective, produced on Machine 2

Assume the sample space is

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

with probabilities

$$
P(\{G_1\})=0.50,\qquad P(\{G_2\})=0.20,\qquad P(\{D_1\})=0.10,\qquad P(\{D_2\})=0.20.
$$

Define two discrete random variables on the same probability space:

* $X$ records only whether the item is good or defective:
  $$
  X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.
  $$

* $Y$ records both quality and machine:
  $$
  Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4.
  $$

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that $\sigma(X)\subseteq \sigma(Y)$, and explain why $Y$ contains more information than $X$.

5. Find the marginal distribution of $X$.

6. Find the marginal distribution of $Y$.

7. Compute the joint probabilities
   $$
   P(X=0,Y=1),\quad P(X=0,Y=2),\quad P(X=1,Y=3),\quad P(X=1,Y=4).
   $$

8. Compute the conditional probabilities
   $$
   P(X=1\mid Y=3),\qquad P(X=1\mid Y=4),\qquad P(Y=3\mid X=1),\qquad P(Y=4\mid X=1).
   $$

9. Interpret the difference between conditioning on $X$ and conditioning on $Y$.

10. Explain, in words, what information is available in $\sigma(X)$ but not in $\sigma(Y)$, and what information is available in $\sigma(Y)$ but not in $\sigma(X)$.

11. Give a Python based visualization of the model.

12. Find the marginal, joint, and conditional Shannon Entropy of $X$ and $Y$.

### Answer

#### 1. Probability space

The probability space is given by

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

$$
\mathfrak F=\mathcal P(\Omega),
$$

with probability measure $P$ defined as

$$
P(\{G_1\})=0.50,\qquad
P(\{G_2\})=0.20,\qquad
P(\{D_1\})=0.10,\qquad
P(\{D_2\})=0.20.
$$

Also,

$$
0.50+0.20+0.10+0.20=1.
$$

Hence this is a valid probability space.

---

#### 2. Sigma-algebra generated by $X$

The random variable $X$ records only the quality type:

$$
X(G_1)=0,\qquad X(G_2)=0,\qquad X(D_1)=1,\qquad X(D_2)=1.
$$

So the inverse images are

$$
X^{-1}(\{0\})=\{G_1,G_2\},
$$

and

$$
X^{-1}(\{1\})=\{D_1,D_2\}.
$$

Therefore,

$$
\sigma(X)=\left\{\varnothing,\Omega,\{G_1,G_2\},\{D_1,D_2\}\right\}.
$$

---

#### 3. Sigma-algebra generated by $Y$

The random variable $Y$ records both the quality and the machine:

$$
Y(G_1)=1,\qquad Y(G_2)=2,\qquad Y(D_1)=3,\qquad Y(D_2)=4.
$$

Each outcome in $\Omega$ is mapped to a different value, so $Y$ separates all points of the sample space. Hence

$$
\sigma(Y)=\mathcal P(\Omega).
$$

Explicitly, this is the full power set of $\Omega$.

---

#### 4. Show that $\sigma(X)\subseteq \sigma(Y)$

Since

$$
\sigma(Y)=\mathcal P(\Omega),
$$

every event in $\sigma(X)$ is automatically contained in $\sigma(Y)$. Therefore,

$$
\sigma(X)\subseteq \sigma(Y).
$$

The variable $X$ only tells whether the item is good or defective, while $Y$ tells both the quality and the machine number. So $Y$ contains more information than $X$.

---

#### 5. Marginal distribution of $X$

For $X=0$,

$$
P(X=0)=P(\{G_1,G_2\})=0.50+0.20=0.70.
$$

For $X=1$,

$$
P(X=1)=P(\{D_1,D_2\})=0.10+0.20=0.30.
$$

So the marginal distribution of $X$ is

$$
P(X=0)=0.70,\qquad P(X=1)=0.30.
$$

---

#### 6. Marginal distribution of $Y$

The marginal probabilities for $Y$ are

$$
P(Y=1)=P(\{G_1\})=0.50,
$$

$$
P(Y=2)=P(\{G_2\})=0.20,
$$

$$
P(Y=3)=P(\{D_1\})=0.10,
$$

$$
P(Y=4)=P(\{D_2\})=0.20.
$$

---

#### 7. Joint probabilities

The required joint probabilities are

$$
P(X=0,Y=1)=P(\{G_1\})=0.50,
$$

$$
P(X=0,Y=2)=P(\{G_2\})=0.20,
$$

$$
P(X=1,Y=3)=P(\{D_1\})=0.10,
$$

$$
P(X=1,Y=4)=P(\{D_2\})=0.20.
$$

All remaining joint probabilities are zero.

---

#### 8. Conditional probabilities

Since $Y=3$ means the outcome is $D_1$, we get

$$
P(X=1\mid Y=3)=1.
$$

Similarly, $Y=4$ means the outcome is $D_2$, so

$$
P(X=1\mid Y=4)=1.
$$

Next,

$$
P(Y=3\mid X=1)
=
\frac{P(Y=3,X=1)}{P(X=1)}
=
\frac{0.10}{0.30}
=
\frac{1}{3}.
$$

Also,

$$
P(Y=4\mid X=1)
=
\frac{P(Y=4,X=1)}{P(X=1)}
=
\frac{0.20}{0.30}
=
\frac{2}{3}.
$$

---

#### 9. Interpretation of conditioning on $X$ and on $Y$

Conditioning on $Y$ gives more detailed information than conditioning on $X$.

For example, if $Y=3$, then the exact outcome is known and the item must be defective. So

$$
P(X=1\mid Y=3)=1.
$$

But if we only know that $X=1$, we know the item is defective but we still do not know whether it came from Machine 1 or Machine 2. Therefore,

$$
P(Y=3\mid X=1)=\frac{1}{3},\qquad
P(Y=4\mid X=1)=\frac{2}{3}.
$$

---

#### 10. Information in $\sigma(X)$ and $\sigma(Y)$

The sigma-algebra $\sigma(X)$ contains only information about whether the item is good or defective. It does not distinguish between Machine 1 and Machine 2.

The sigma-algebra $\sigma(Y)$ contains complete information about the outcome, since it tells both the quality and the machine. Therefore, $\sigma(Y)$ is finer than $\sigma(X)$.

### Entropy

Using natural logarithms, the Shannon entropy of a discrete random variable $Z$ is

$$
H(Z)=-\sum_z P(z)\ln P(z).
$$

For $X$,

$$
H(X)=-(0.70\ln 0.70+0.30\ln 0.30).
$$

Therefore,

$$
H(X)\approx 0.61086.
$$

For $Y$,

$$
H(Y)=-(0.50\ln 0.50+0.20\ln 0.20+0.10\ln 0.10+0.20\ln 0.20).
$$

Hence,

$$
H(Y)\approx 1.22061.
$$

For the joint entropy,

$$
H(X,Y)=-\sum_{x,y}P(x,y)\ln P(x,y).
$$

Only the nonzero joint probabilities contribute, so

$$
H(X,Y)=-(0.50\ln 0.50+0.20\ln 0.20+0.10\ln 0.10+0.20\ln 0.20).
$$

Thus,

$$
H(X,Y)\approx 1.22061.
$$

Since each value of $Y$ determines $X$ exactly,

$$
H(X\mid Y)=0.
$$

Also,

$$
H(Y\mid X)=H(X,Y)-H(X).
$$

So,

$$
H(Y\mid X)\approx 1.22061-0.61086=0.60975.
$$

Therefore, using natural logarithms,

$$
H(X)\approx 0.61086,
$$

$$
H(Y)\approx 1.22061,
$$

$$
H(X,Y)\approx 1.22061,
$$

$$
H(X\mid Y)=0,
$$

$$
H(Y\mid X)\approx 0.60975.
$$
