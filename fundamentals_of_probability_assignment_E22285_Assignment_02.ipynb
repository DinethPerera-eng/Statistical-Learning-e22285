#Q3) Transformer Failiures

In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.
2. Define $X$ as a random variable on this space.
3. Determine the probability mass function of $X$.
4. Compute the expectation $E[X]$.
5. Compute the variance $\mathrm{Var}(X)$.
6. Find $P(X\ge 1)$ and $P(X\ge 2)$.
7. Briefly explain the practical meaning of the expectation and variance in this setting.



### Answer

#### 1. Probability space

For this model, take

$$
\Omega=\{0,1,2,3\},
$$

and let

$$
\mathfrak F=\mathfrak P(\Omega),
$$

where $\mathfrak P(\Omega)$ is the power set of $\Omega$.

The probability measure $P$ is assigned by

$$
P(\{0\})=0.50,\qquad
P(\{1\})=0.30,\qquad
P(\{2\})=0.15,\qquad
P(\{3\})=0.05.
$$

For any event $A\subseteq\Omega$,

$$
P(A)=\sum_{\omega\in A}P(\{\omega\}).
$$

---

#### 2. Definition of the random variable

Define

$$
X:\Omega\to\mathbb R
$$

by

$$
X(\omega)=\omega,\qquad \omega\in\Omega.
$$

Thus, $X$ gives the number of emergency transformer failures reported in one day.

---

#### 3. Probability mass function

The probability mass function of $X$ is

$$
p_X(x)=
\begin{cases}
0.50, & x=0,\\
0.30, & x=1,\\
0.15, & x=2,\\
0.05, & x=3,\\
0, & \text{otherwise}.
\end{cases}
$$

---

#### 4. Expectation

$$
E[X]
=
\sum_x x\,p_X(x)
$$

$$
=
0(0.50)+1(0.30)+2(0.15)+3(0.05)
$$

$$
=0.75.
$$

---

#### 5. Variance

First,

$$
E[X^2]
=
0^2(0.50)+1^2(0.30)+2^2(0.15)+3^2(0.05)
$$

$$
=1.35.
$$

Therefore,

$$
\mathrm{Var}(X)
=
E[X^2]-(E[X])^2
$$

$$
=
1.35-(0.75)^2
$$

$$
=0.7875.
$$

---

#### 6. Required probabilities

$$
P(X\geq 1)
=
P(X=1)+P(X=2)+P(X=3)
$$

$$
=
0.30+0.15+0.05
=
0.50.
$$

Also,

$$
P(X\geq 2)
=
P(X=2)+P(X=3)
$$

$$
=
0.15+0.05
=
0.20.
$$

---

#### 7. Practical meaning

The expectation

$$
E[X]=0.75
$$

means that the average number of emergency transformer failures per day is $0.75$.

The variance

$$
\mathrm{Var}(X)=0.7875
$$

describes how much the daily number of failures varies around that average.

# Q4) Probability of Defective Items

In a manufacturing process, the proportion of defective items in a large batch produced on a given day is modeled by a continuous random variable $X$. Since $X$ represents a proportion, it takes values in the interval $[0,1]$. Suppose that $X$ has a Beta distribution with parameters $\alpha=2$ and $\beta=18$, so that its density is

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},
\end{cases}
$$

where $B(2,18)$ is the Beta function.

Answer the following:

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

2. Define $X$ as a continuous random variable on this space.

3. Write down the probability density function of $X$.

4. Verify that $f$ is a valid probability density function.

5. Compute the expectation $E[X]$.

6. Compute the variance $\mathrm{Var}(X)$.

7. Find the probability that the defective proportion is at most $0.10$, that is, compute

$$
P(X\le 0.10).
$$

You may leave your answer in Beta integral form.

8. Find the probability that the defective proportion lies between $0.05$ and $0.15$, that is, compute

$$
P(0.05\le X\le 0.15).
$$

You may leave your answer in integral form.

9. Briefly explain why a Beta distribution is a reasonable model for this situation.

10. Interpret the expectation and variance in the context of production quality.


### Answer

#### 1. Probability space

A suitable probability space for this model is

$$
\Omega=[0,1], \qquad \mathfrak F=\mathfrak B([0,1]),
$$

where $\mathfrak B([0,1])$ is the Borel $\sigma$-algebra on $[0,1]$.

For any event $A\in\mathfrak F$, define

$$
P(A)=\int_A f(x)\,dx,
$$

where

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1,\\
0, & \text{otherwise}.
\end{cases}
$$

---

#### 2. Definition of the random variable

Define the random variable

$$
X:\Omega\to\mathbb R
$$

by

$$
X(\omega)=\omega,\qquad \omega\in[0,1].
$$

Thus, $X$ represents the defective proportion in a batch.

---

#### 3. Probability density function

The probability density function of $X$ is

$$
f_X(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1,\\
0, & \text{otherwise}.
\end{cases}
$$

---

#### 4. Verification of the density function

For every $x$, we have

$$
f(x)\geq 0.
$$

Also,

$$
\int_{-\infty}^{\infty} f(x)\,dx
=
\int_0^1 \dfrac{1}{B(2,18)}x(1-x)^{17}\,dx
$$

$$
=
\dfrac{1}{B(2,18)}B(2,18)
=
1.
$$

Therefore, $f$ is a valid probability density function.

---

#### 5. Expectation

Since

$$
X\sim \mathrm{Beta}(2,18),
$$

the expectation is

$$
E[X]=\frac{\alpha}{\alpha+\beta}
=
\frac{2}{2+18}
=
\frac{2}{20}
=
0.10.
$$

---

#### 6. Variance

For a Beta$(\alpha,\beta)$ random variable,

$$
\mathrm{Var}(X)
=
\frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}.
$$

Therefore,

$$
\mathrm{Var}(X)
=
\frac{2\cdot18}{20^2\cdot21}
=
\frac{36}{8400}
=
\frac{3}{700}
\approx 0.00429.
$$

---

#### 7. Probability that the defective proportion is at most $0.10$

$$
P(X\leq 0.10)
=
\int_0^{0.10}
\dfrac{1}{B(2,18)}x(1-x)^{17}\,dx.
$$

Equivalently,

$$
P(X\leq 0.10)=I_{0.10}(2,18).
$$

---

#### 8. Probability that the defective proportion lies between $0.05$ and $0.15$

$$
P(0.05\leq X\leq 0.15)
=
\int_{0.05}^{0.15}
\dfrac{1}{B(2,18)}x(1-x)^{17}\,dx.
$$

Equivalently,

$$
P(0.05\leq X\leq 0.15)
=
I_{0.15}(2,18)-I_{0.05}(2,18).
$$

---

#### 9. Reason for using the Beta distribution

A Beta distribution is suitable here because the defective proportion is a continuous value restricted to the interval $[0,1]$. The Beta family is commonly used to model proportions and rates.

---

#### 10. Interpretation of expectation and variance

The value

$$
E[X]=0.10
$$

means that the long-run average defective proportion is $10\%$.

The variance

$$
\mathrm{Var}(X)\approx 0.00429
$$

shows how much the defective proportion changes from batch to batch, and it reflects the stability of production quality over time.

### The plot of the $\beta$-distribution

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import beta

def plot_beta_distribution(alpha, beta_val):
    """
    Plots the Beta distribution PDF for given alpha and beta parameters using Plotly.
    """
    # Create a range of x values from 0 to 1
    x = np.linspace(0, 1, 1000)

    # Calculate the PDF values for the Beta distribution
    y = beta.pdf(x, alpha, beta_val)

    # Create the Plotly figure
    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=x,
        y=y,
        mode='lines',
        name=f'Beta($\\alpha={alpha}, \\beta={beta_val}$)',
        line=dict(width=2)
    ))

    # Update layout for better readability
    fig.update_layout(
        title=f'Beta Distribution Density Function ($\\alpha={alpha}, \\beta={beta_val}$)',
        xaxis_title='x',
        yaxis_title='PDF(x)',
        template='plotly_white'
    )

    # To display the plot, use fig.show()
    return fig

# Example usage:
fig = plot_beta_distribution(alpha=2, beta_val=18)
fig.show()
```

# Q5) Expected Project Completion Time

A construction company is estimating the time required to install a backup generator at a hospital. Let $X$ denote the number of hours needed to complete the installation. Based on expert judgment, the shortest possible time is $4$ hours, the longest possible time is $10$ hours, and the most likely completion time is $6$ hours. Assume that $X$ follows a Triangular distribution with parameters $a=4$, $b=10$, and $c=6$.

Answer the following:

1. Specify the support of $X$.

2. Write down the probability density function $f_X(x)$.

3. Verify that $f_X$ integrates to $1$ over its support.

4. Compute the expectation $E[X]$.

5. Compute the variance $\mathrm{Var}(X)$.

6. Interpret the values of $a$, $b$, and $c$ in the context of this problem.

7. Explain briefly why a Triangular distribution is a reasonable model in this setting.

8. Compute the probability that the installation takes at most $5$ hours, that is, compute

$$
P(X \leq 5).
$$

9. Compute the probability that the installation takes between $5$ and $8$ hours, that is, compute

$$
P(5 \leq X \leq 8).
$$

10. Briefly explain how this model could be used in project planning or sensitivity analysis.


### Answer

#### 1. Support of $X$

The installation time cannot be less than $4$ hours or more than $10$ hours. Therefore, the support of $X$ is

$$
4\leq X\leq 10.
$$

---

#### 2. Probability density function

Since

$$
X\sim \mathrm{Triangular}(4,10,6),
$$

the density function is

$$
f_X(x)=
\begin{cases}
\dfrac{2(x-4)}{(10-4)(6-4)}, & 4\leq x<6,\\
\dfrac{2}{10-4}, & x=6,\\
\dfrac{2(10-x)}{(10-4)(10-6)}, & 6<x\leq 10,\\
0, & \text{otherwise}.
\end{cases}
$$

After simplifying,

$$
f_X(x)=
\begin{cases}
\dfrac{x-4}{6}, & 4\leq x<6,\\
\dfrac{1}{3}, & x=6,\\
\dfrac{10-x}{12}, & 6<x\leq 10,\\
0, & \text{otherwise}.
\end{cases}
$$

---

#### 3. Verification that $f_X$ is a valid density

To check that the density integrates to $1$,

$$
\int_4^6 \frac{x-4}{6}\,dx+\int_6^{10}\frac{10-x}{12}\,dx.
$$

For the first interval,

$$
\int_4^6 \frac{x-4}{6}\,dx
=
\frac{1}{6}\int_4^6 (x-4)\,dx
=
\frac{1}{6}\cdot 2
=
\frac{1}{3}.
$$

For the second interval,

$$
\int_6^{10}\frac{10-x}{12}\,dx
=
\frac{1}{12}\int_6^{10}(10-x)\,dx
=
\frac{1}{12}\cdot 8
=
\frac{2}{3}.
$$

Hence,

$$
\int_{-\infty}^{\infty}f_X(x)\,dx
=
\frac{1}{3}+\frac{2}{3}
=
1.
$$

Therefore, $f_X$ is a valid probability density function.

---

#### 4. Expectation

For a Triangular distribution,

$$
E[X]=\frac{a+b+c}{3}.
$$

Substituting $a=4$, $b=10$, and $c=6$,

$$
E[X]=\frac{4+10+6}{3}
=
\frac{20}{3}
\approx 6.67.
$$

---

#### 5. Variance

The variance of a Triangular distribution is

$$
\mathrm{Var}(X)
=
\frac{a^2+b^2+c^2-ab-ac-bc}{18}.
$$

Therefore,

$$
\mathrm{Var}(X)
=
\frac{4^2+10^2+6^2-(4)(10)-(4)(6)-(10)(6)}{18}.
$$

So,

$$
\mathrm{Var}(X)
=
\frac{16+100+36-40-24-60}{18}
=
\frac{28}{18}
=
\frac{14}{9}
\approx 1.56.
$$

---

#### 6. Meaning of the parameters

In this problem,

* $a=4$ is the minimum possible installation time,
* $b=10$ is the maximum possible installation time,
* $c=6$ is the most likely installation time.

---

#### 7. Reason for using a Triangular distribution

A Triangular distribution is suitable here because the company has expert estimates for the minimum, maximum, and most likely completion times, even if there is not enough historical data for a more detailed model.

---

#### 8. Probability that the installation takes at most $5$ hours

Since $5$ lies in the interval $[4,6]$, use the first part of the density:

$$
P(X\leq 5)
=
\int_4^5 \frac{x-4}{6}\,dx.
$$

Thus,

$$
P(X\leq 5)
=
\frac{1}{6}\int_4^5(x-4)\,dx
=
\frac{1}{6}\cdot\frac{1}{2}
=
\frac{1}{12}.
$$

Therefore,

$$
P(X\leq 5)=\frac{1}{12}\approx 0.0833.
$$

---

#### 9. Probability that the installation takes between $5$ and $8$ hours

Since the interval crosses the mode $x=6$, split the probability into two parts:

$$
P(5\leq X\leq 8)
=
\int_5^6 \frac{x-4}{6}\,dx
+
\int_6^8 \frac{10-x}{12}\,dx.
$$

For the first part,

$$
\int_5^6 \frac{x-4}{6}\,dx
=
\frac{1}{6}\int_5^6(x-4)\,dx
=
\frac{1}{6}\cdot\frac{3}{2}
=
\frac{1}{4}.
$$

For the second part,

$$
\int_6^8 \frac{10-x}{12}\,dx
=
\frac{1}{12}\int_6^8(10-x)\,dx
=
\frac{1}{12}\cdot 6
=
\frac{1}{2}.
$$

Therefore,

$$
P(5\leq X\leq 8)
=
\frac{1}{4}+\frac{1}{2}
=
\frac{3}{4}.
$$

---

#### 10. Use in project planning

This model can help estimate the chance of finishing the installation within a given time period. It can also support sensitivity analysis by showing how uncertainty in task duration may affect project schedules, staffing, and contingency planning.

### Plot of the triangular distribution

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang

# Parameters given
a = 4
b = 10
c_mode = 6

# Scipy parametrization
# loc is the minimum (a)
# scale is the width (b - a)
# c_scipy is the ratio of the mode relative to the width (c - a) / (b - a)
loc = a
scale = b - a
c_scipy = (c_mode - a) / (b - a)

# Generate data points
x = np.linspace(0, b, 500)
y = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)

# Create the Plotly figure
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode='lines',
    fill='tozeroy',
    name='Triangular Distribution',
    line=dict(color='royalblue', width=2)
))

# Layout updates
fig.update_layout(
    title=f'Triangular Distribution (a={a}, b={b}, c={c_mode})',
    xaxis_title='Time (hours)',
    yaxis_title='Probability Density',
    template='plotly_white'
)

# Display the plot
fig.show()
```

#Q6) Expected Project Completion Time Using PERT

A project manager is estimating the time required to complete a critical software integration task. Let $X$ denote the number of days required to complete the task. Based on expert judgment, the shortest possible time is $a=8$ days, the longest possible time is $b=20$ days, and the most likely completion time is $c=11$ days. Assume that $X$ follows a PERT distribution with parameters $(a,b,c)$.

The PERT distribution is defined as a Beta distribution scaled to the interval $[a,b]$, with internal shape parameters

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right).
$$

Its probability density function is

$$
f(x)=\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}}, \qquad a \leq x \leq b,
$$

where

$$
B(\alpha,\beta)=\int_0^1 t^{\alpha-1}(1-t)^{\beta-1},dt.
$$

Answer the following:

1. Compute the values of the shape parameters $\alpha$ and $\beta$.

2. Specify the support of $X$.

3. Write down the probability density function $f_X(x)$ for this model.

4. State the probability space $(\Omega,\mathfrak F,P)$ suitable for this continuous model.

5. Define $X$ as a random variable on this space.

6. Show that the expectation $E[X]$ is given by

$$
E[X]=\frac{a+4c+b}{6}.
$$

7. Compute the variance $\mathrm{Var}(X)$ using the approximation

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}.
$$

8. Compute the approximate standard deviation $\sigma$.

9. Briefly explain why a PERT distribution may be more appropriate than a Triangular distribution in this setting byt comparing the two results.

10. Interpret the practical meaning of the values $a$, $b$, $c$, $E[X]$, and $\sigma$ in the context of project planning.


### Answer

#### 1. Shape parameters

The given PERT parameters are

$$
a=8,\qquad b=20,\qquad c=11.
$$

Using

$$
\alpha=1+4\left(\frac{c-a}{b-a}\right),
$$

we get

$$
\alpha
=
1+4\left(\frac{11-8}{20-8}\right)
=
1+4\left(\frac{3}{12}\right)
=
2.
$$

Similarly,

$$
\beta=1+4\left(\frac{b-c}{b-a}\right),
$$

so

$$
\beta
=
1+4\left(\frac{20-11}{20-8}\right)
=
1+4\left(\frac{9}{12}\right)
=
4.
$$

Therefore,

$$
\alpha=2,\qquad \beta=4.
$$

---

#### 2. Support of $X$

Since the task duration can vary from the shortest possible time of $8$ days to the longest possible time of $20$ days, the support is

$$
8\leq X\leq 20.
$$

---

#### 3. Probability density function

The PERT density is

$$
f_X(x)
=
\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}
{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}},
\qquad a\leq x\leq b.
$$

Substituting

$$
a=8,\qquad b=20,\qquad \alpha=2,\qquad \beta=4,
$$

gives

$$
f_X(x)
=
\frac{(x-8)(20-x)^3}{B(2,4)\,12^5},
\qquad 8\leq x\leq 20.
$$

Also,

$$
B(2,4)
=
\frac{\Gamma(2)\Gamma(4)}{\Gamma(6)}
=
\frac{1!\,3!}{5!}
=
\frac{1}{20}.
$$

Therefore,

$$
f_X(x)
=
\frac{20(x-8)(20-x)^3}{12^5},
\qquad 8\leq x\leq 20,
$$

and

$$
f_X(x)=0
\qquad \text{otherwise}.
$$

---

#### 4. Probability space

A suitable probability space is

$$
\Omega=[8,20],\qquad \mathfrak F=\mathfrak B([8,20]),
$$

where $\mathfrak B([8,20])$ is the Borel $\sigma$-algebra on $[8,20]$.

For any event $A\in\mathfrak F$, define

$$
P(A)=\int_A f_X(x)\,dx.
$$

---

#### 5. Definition of the random variable

Define

$$
X:\Omega\to\mathbb R
$$

by

$$
X(\omega)=\omega,\qquad \omega\in[8,20].
$$

Thus, $X$ gives the number of days required to complete the software integration task.

---

#### 6. Expectation

For a scaled Beta random variable on $[a,b]$,

$$
E[X]
=
a+(b-a)\frac{\alpha}{\alpha+\beta}.
$$

Here,

$$
\alpha+\beta
=
\left(1+4\frac{c-a}{b-a}\right)
+
\left(1+4\frac{b-c}{b-a}\right)
=
6.
$$

Therefore,

$$
E[X]
=
a+(b-a)\frac{1+4\frac{c-a}{b-a}}{6}.
$$

Simplifying,

$$
E[X]
=
a+\frac{b-a}{6}+\frac{4(c-a)}{6}
$$

$$
=
\frac{a+4c+b}{6}.
$$

Now substitute the given values:

$$
E[X]
=
\frac{8+4(11)+20}{6}
=
\frac{72}{6}
=
12.
$$

Hence,

$$
E[X]=12.
$$

---

#### 7. Variance

Using the given approximation,

$$
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7}.
$$

Substituting

$$
E[X]=12,\qquad a=8,\qquad b=20,
$$

we get

$$
\mathrm{Var}(X)
=
\frac{(12-8)(20-12)}{7}
=
\frac{4\cdot 8}{7}
=
\frac{32}{7}.
$$

Therefore,

$$
\mathrm{Var}(X)\approx 4.57.
$$

---

#### 8. Standard deviation

The standard deviation is

$$
\sigma
=
\sqrt{\mathrm{Var}(X)}
=
\sqrt{\frac{32}{7}}
\approx 2.14.
$$

---

#### 9. Comparison with the Triangular distribution

For the same values

$$
a=8,\qquad b=20,\qquad c=11,
$$

the Triangular distribution gives

$$
E[X]_{\text{Triangular}}=\frac{8+20+11}{3}=13,
$$

and

$$
\mathrm{Var}(X)_{\text{Triangular}}=\frac{49}{6}\approx 8.17.
$$

For the PERT model,

$$
E[X]_{\text{PERT}}=12,
$$

and

$$
\mathrm{Var}(X)_{\text{PERT}}=\frac{32}{7}\approx 4.57.
$$

The PERT distribution gives more weight to the most likely time and produces a smoother shape with lower variability than the Triangular distribution. Therefore, it may be more suitable when expert judgment suggests that the most likely value should have stronger influence.

---

#### 10. Practical interpretation

In this task:

* $a=8$ is the shortest possible completion time,
* $b=20$ is the longest possible completion time,
* $c=11$ is the most likely completion time,
* $E[X]=12$ is the expected number of days needed,
* $\sigma\approx 2.14$ measures the spread or uncertainty in the completion time.

These values help the project manager estimate a realistic task duration and understand how much variation may occur around the expected completion time.

### Plot of the triangle and PERT distributions

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang, beta

# Parameters
a = 4
b = 10
c_mode = 6

# --- Triangular Distribution Calculation ---
# Scipy parametrization for Triangular
c_scipy = (c_mode - a) / (b - a)
loc = a
scale = b - a

# --- PERT Distribution Calculation ---
# PERT is a Beta distribution with specific alpha and beta parameters
# alpha = 1 + 4 * (c - a) / (b - a)
# beta = 1 + 4 * (b - c) / (b - a)
alpha_pert = 1 + 4 * (c_mode - a) / (b - a)
beta_pert = 1 + 4 * (b - c_mode) / (b - a)

# Generate data points
x = np.linspace(0, b, 500)
y_triang = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)
y_pert = beta.pdf(x, alpha_pert, beta_pert, loc=a, scale=b-a)

# Create the Plotly figure
fig = go.Figure()

# Add Triangular Distribution
fig.add_trace(go.Scatter(
    x=x,
    y=y_triang,
    mode='lines',
    name='Triangular Distribution',
    line=dict(color='royalblue', width=2, dash='dash')
))

# Add PERT Distribution
fig.add_trace(go.Scatter(
    x=x,
    y=y_pert,
    mode='lines',
    name='PERT Distribution',
    line=dict(color='firebrick', width=2)
))

# Layout updates
fig.update_layout(
    title=f'Comparison: Triangular vs PERT (a={a}, b={b}, c={c_mode})',
    xaxis_title='Time (hours)',
    yaxis_title='Probability Density',
    template='plotly_white'
)

# Display the plot
fig.show()
```

#Q7) If the PERT Expectation is to Match Our Intuition

A project manager models the duration $X$ of a task by a PERT distribution with parameters $a$, $b$, and $c$, where $a$ is the minimum time, $b$ is the maximum time, and $c$ is the most likely time.

Suppose the minimum time is $a=5$ days and the most likely time is $c=11$ days. If the expected time computed from the PERT model is also to be the same as the most likely time:

1. Determine the value of the maximum time $b$.

2. What happens to the variance?

### Answer

#### 1. Finding the maximum time $b$

For a PERT distribution,

$$
E[X]=\frac{a+4c+b}{6}.
$$

Since the expected time is required to be equal to the most likely time,

$$
E[X]=c.
$$

Therefore,

$$
\frac{a+4c+b}{6}=c.
$$

Multiplying by $6$,

$$
a+4c+b=6c.
$$

So,

$$
a+b=2c,
$$

and hence,

$$
b=2c-a.
$$

Substituting $a=5$ and $c=11$,

$$
b=2(11)-5=22-5=17.
$$

Therefore,

$$
\boxed{b=17}.
$$

---

#### 2. Effect on the variance

Using the PERT variance approximation,

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}.
$$

Since $E[X]=c$,

$$
\mathrm{Var}(X)=\frac{(c-a)(b-c)}{7}.
$$

Also, from part 1,

$$
b=2c-a.
$$

Therefore,

$$
b-c=c-a.
$$

Hence,

$$
\mathrm{Var}(X)=\frac{(c-a)^2}{7}.
$$

Equivalently,

$$
\mathrm{Var}(X)=\frac{(b-a)^2}{28}.
$$

For the given values,

$$
\mathrm{Var}(X)=\frac{(11-5)^2}{7}
=
\frac{36}{7}.
$$

When the expected value is equal to the most likely value, the PERT model becomes symmetric.
