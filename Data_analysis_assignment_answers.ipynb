## Part A – Question 1

### Bias of the MLE Covariance Matrix

The Maximum Likelihood Estimator of the population covariance matrix $\boldsymbol{\Sigma}$ is

$$
\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}
=
\frac{1}{n}\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T
$$

where

$$
\widehat{\boldsymbol{\mu}}_n
=
\frac{1}{n}\sum_{i=1}^{n}\mathbf{X}_i
$$

is the empirical sample mean.

To find the bias, rewrite the difference from the sample mean using the true population mean $\boldsymbol{\mu}$:

$$
\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n
=
(\mathbf{X}_i-\boldsymbol{\mu})
-
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
$$

Substituting into the covariance summation gives

$$
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T
=
\sum_{i=1}^{n}
\left[
(\mathbf{X}_i-\boldsymbol{\mu})
-
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
\right]
\left[
(\mathbf{X}_i-\boldsymbol{\mu})
-
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
\right]^T
$$

After expanding,

$$
=
\sum_{i=1}^{n}
(\mathbf{X}_i-\boldsymbol{\mu})
(\mathbf{X}_i-\boldsymbol{\mu})^T
-
\sum_{i=1}^{n}
(\mathbf{X}_i-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T
-
\sum_{i=1}^{n}
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\mathbf{X}_i-\boldsymbol{\mu})^T
+
\sum_{i=1}^{n}
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T
$$

Since

$$
\sum_{i=1}^{n}(\mathbf{X}_i-\boldsymbol{\mu})
=
n(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu}),
$$

the expression becomes

$$
=
\sum_{i=1}^{n}
(\mathbf{X}_i-\boldsymbol{\mu})
(\mathbf{X}_i-\boldsymbol{\mu})^T
-
n(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T
$$

Now take expectation:

$$
\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
=
\frac{1}{n}
\sum_{i=1}^{n}
\mathbb{E}
[
(\mathbf{X}_i-\boldsymbol{\mu})
(\mathbf{X}_i-\boldsymbol{\mu})^T
]
-
\mathbb{E}
[
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T
]
$$

Using

$$
\mathbb{E}[(\mathbf{X}_i-\boldsymbol{\mu})(\mathbf{X}_i-\boldsymbol{\mu})^T]
=
\boldsymbol{\Sigma}
$$

and

$$
\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T]
=
\mathrm{Var}(\widehat{\boldsymbol{\mu}}_n)
=
\frac{1}{n}\boldsymbol{\Sigma},
$$

we get

$$
\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
=
\frac{1}{n}(n\boldsymbol{\Sigma})
-
\frac{1}{n}\boldsymbol{\Sigma}
$$

$$
=
\boldsymbol{\Sigma}
-
\frac{1}{n}\boldsymbol{\Sigma}
$$

$$
=
\left(\frac{n-1}{n}\right)\boldsymbol{\Sigma}
$$

Therefore,

$$
\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
=
\frac{n-1}{n}\boldsymbol{\Sigma}
$$

Since

$$
\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
\neq
\boldsymbol{\Sigma},
$$

the MLE covariance estimator is biased downward in finite samples.

---

### Applying Bessel’s Correction

To remove the bias, multiply the MLE estimator by

$$
\frac{n}{n-1}
$$

So,

$$
\mathbf{S}
=
\frac{n}{n-1}
\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}
$$

Substituting the MLE formula,

$$
\mathbf{S}
=
\frac{n}{n-1}
\left(
\frac{1}{n}
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T
\right)
$$

Therefore,

$$
\mathbf{S}
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T
$$

Now check the expectation:

$$
\mathbb{E}[\mathbf{S}]
=
\frac{n}{n-1}
\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
$$

$$
=
\frac{n}{n-1}
\left(
\frac{n-1}{n}
\right)
\boldsymbol{\Sigma}
$$

$$
=
\boldsymbol{\Sigma}
$$

Hence,

$$
\mathbb{E}[\mathbf{S}]
=
\boldsymbol{\Sigma}
$$

Therefore, $\mathbf{S}$ is the unbiased covariance estimator obtained using Bessel’s correction.

### Slutsky’s Theorem and Asymptotic Justification

Let $\{X_n\}$ and $\{Y_n\}$ be sequences of random variables or random vectors. Suppose that

$$
X_n \xrightarrow{d} X
$$

and

$$
Y_n \xrightarrow{p} c
$$

where $c$ is a constant scalar or matrix. Then Slutsky’s Theorem gives the following results:

$$
X_n + Y_n \xrightarrow{d} X + c
$$

$$
X_nY_n \xrightarrow{d} cX
$$

$$
\frac{X_n}{Y_n} \xrightarrow{d} \frac{X}{c}, \quad c \neq 0
$$

---

### Operational Distribution Justification

According to the Lindeberg-Lévy Multivariate Central Limit Theorem,

$$
\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})
\xrightarrow{d}
\mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})
$$

This shows that the scaled difference between the sample mean and the population mean converges in distribution to a multivariate normal distribution with mean vector $\mathbf{0}$ and covariance matrix $\boldsymbol{\Sigma}$.

To standardize this limiting distribution, multiply by the inverse square root of the population covariance matrix:

$$
\boldsymbol{\Sigma}^{-1/2}
\left[
\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})
\right]
\xrightarrow{d}
\mathscr{N}(\mathbf{0}, \mathbf{I})
$$

By the Weak Law of Large Numbers, the unbiased sample covariance matrix $\mathbf{S}$ is a consistent estimator of the population covariance matrix $\boldsymbol{\Sigma}$. Therefore,

$$
\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma}
$$

Since the inverse square root operation is continuous for positive definite covariance matrices,

$$
\mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}
$$

Now, applying Slutsky’s Theorem,

$$
\mathbf{S}^{-1/2}
\left[
\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})
\right]
\xrightarrow{d}
\boldsymbol{\Sigma}^{-1/2}
\mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})
$$

Therefore,

$$
\mathbf{S}^{-1/2}
\left[
\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})
\right]
\xrightarrow{d}
\mathscr{N}(\mathbf{0}, \mathbf{I})
$$

This means that replacing the unknown population covariance matrix $\boldsymbol{\Sigma}$ with the sample covariance matrix $\mathbf{S}$ does not change the asymptotic distribution.

Hence, for large sample sizes,

$$
\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})
\sim
\mathscr{N}(\mathbf{0}, \mathbf{S})
$$

Equivalently,

$$
\widehat{\boldsymbol{\mu}}_n
\sim
\mathscr{N}
\left(
\boldsymbol{\mu},
\frac{1}{n}\mathbf{S}
\right)
$$

Therefore, using $\mathbf{S}$ instead of $\boldsymbol{\Sigma}$ is asymptotically valid, because $\mathbf{S}$ converges in probability to $\boldsymbol{\Sigma}$ as the sample size becomes large.

````python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015
base_data[4000:, 2] -= 0.010

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------


def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g consecutive chunks and tests first-moment
    homogeneity using Wilks' Lambda and Bartlett's Chi-square approximation.
    """

    X = df.select_dtypes(include=[np.number]).to_numpy(dtype=float)

    n, m = X.shape

    if n % g_chunks != 0:
        raise ValueError("The number of observations must be divisible by g_chunks.")

    block_size = n // g_chunks
    grand_mean = np.mean(X, axis=0)

    W = np.zeros((m, m))
    B = np.zeros((m, m))

    for j in range(g_chunks):
        start = j * block_size
        end = (j + 1) * block_size

        block = X[start:end, :]
        block_mean = np.mean(block, axis=0)

        centered_block = block - block_mean
        W += centered_block.T @ centered_block

        mean_difference = (block_mean - grand_mean).reshape(-1, 1)
        B += block_size * (mean_difference @ mean_difference.T)

    wilks_lambda = np.linalg.det(W) / np.linalg.det(W + B)

    chi_square_calc = -(
        n - 1 - ((m + g_chunks) / 2)
    ) * np.log(wilks_lambda)

    degrees_of_freedom = m * (g_chunks - 1)

    p_value = stats.chi2.sf(chi_square_calc, degrees_of_freedom)

    if p_value <= 0.05:
        conclusion = "Reject H0: the baseline center has shifted over time."
    else:
        conclusion = "Fail to reject H0: the first moment is homogeneous."

    return {
        "Wilks Lambda": wilks_lambda,
        "Bartlett Chi-square": chi_square_calc,
        "Degrees of Freedom": degrees_of_freedom,
        "p-value": p_value,
        "Conclusion": conclusion
    }


results = verify_first_moment_homogeneity(df_strain, g_chunks=5)

for key, value in results.items():
    print(f"{key}: {value}")
````

## 1. Covariance of the Transformed Vector

Let the centered vector be

$$
\widetilde{\mathbf{X}}_i = \mathbf{X}_i - \boldsymbol{\mu}
$$

and the covariance matrix be

$$
\boldsymbol{\Sigma}
=
\mathbb{E}
[
\widetilde{\mathbf{X}}_i
\widetilde{\mathbf{X}}_i^T
]
$$

The spectral decomposition of $\boldsymbol{\Sigma}$ is

$$
\boldsymbol{\Sigma}
=
\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T
$$

where $\mathbf{P}$ is an orthogonal matrix. Therefore,

$$
\mathbf{P}^T\mathbf{P}
=
\mathbf{I}
$$

The transformed vector is

$$
\mathbf{Z}_i
=
\mathbf{P}^T\widetilde{\mathbf{X}}_i
$$

Now calculate the covariance matrix of $\mathbf{Z}_i$:

$$
\mathbb{E}
[
\mathbf{Z}_i\mathbf{Z}_i^T
]
=
\mathbb{E}
[
(\mathbf{P}^T\widetilde{\mathbf{X}}_i)
(\mathbf{P}^T\widetilde{\mathbf{X}}_i)^T
]
$$

Since

$$
(\mathbf{P}^T\widetilde{\mathbf{X}}_i)^T
=
\widetilde{\mathbf{X}}_i^T\mathbf{P}
$$

we get

$$
\mathbb{E}
[
\mathbf{Z}_i\mathbf{Z}_i^T
]
=
\mathbb{E}
[
\mathbf{P}^T
\widetilde{\mathbf{X}}_i
\widetilde{\mathbf{X}}_i^T
\mathbf{P}
]
$$

Since $\mathbf{P}$ is constant,

$$
\mathbb{E}
[
\mathbf{Z}_i\mathbf{Z}_i^T
]
=
\mathbf{P}^T
\mathbb{E}
[
\widetilde{\mathbf{X}}_i
\widetilde{\mathbf{X}}_i^T
]
\mathbf{P}
$$

Substitute

$$
\mathbb{E}
[
\widetilde{\mathbf{X}}_i
\widetilde{\mathbf{X}}_i^T
]
=
\boldsymbol{\Sigma}
$$

Then,

$$
\mathbb{E}
[
\mathbf{Z}_i\mathbf{Z}_i^T
]
=
\mathbf{P}^T
\boldsymbol{\Sigma}
\mathbf{P}
$$

Now substitute

$$
\boldsymbol{\Sigma}
=
\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T
$$

So,

$$
\mathbb{E}
[
\mathbf{Z}_i\mathbf{Z}_i^T
]
=
\mathbf{P}^T
(
\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T
)
\mathbf{P}
$$

Using

$$
\mathbf{P}^T\mathbf{P}
=
\mathbf{I}
$$

we get

$$
\mathbb{E}
[
\mathbf{Z}_i\mathbf{Z}_i^T
]
=
\mathbf{I}\boldsymbol{\Lambda}\mathbf{I}
$$

Therefore,

$$
\mathbb{E}
[
\mathbf{Z}_i\mathbf{Z}_i^T
]
=
\boldsymbol{\Lambda}
$$

This proves that the covariance matrix of the transformed vector is diagonal.

Since $\boldsymbol{\Lambda}$ is diagonal, the cross-covariances between different transformed coordinates are zero. Therefore, for $j \neq k$,

$$
\mathrm{Cov}(Z_{i,j}, Z_{i,k}) = 0
$$

Geometrically, PCA rotates the original coordinate system into orthogonal principal directions. Statistically, this means the transformed principal component coordinates are uncorrelated. Each principal direction carries its own variance independently, and the variance of the $j$-th principal component is $\lambda_j$.

## 2. Total Variance Preservation

The total variance in the original coordinate space is

$$
\mathrm{tr}(\boldsymbol{\Sigma})
$$

Using the spectral decomposition,

$$
\boldsymbol{\Sigma}
=
\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T
$$

Therefore,

$$
\mathrm{tr}(\boldsymbol{\Sigma})
=
\mathrm{tr}
(
\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T
)
$$

Using the cyclic property of trace,

$$
\mathrm{tr}
(
\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T
)
=
\mathrm{tr}
(
\boldsymbol{\Lambda}\mathbf{P}^T\mathbf{P}
)
$$

Since

$$
\mathbf{P}^T\mathbf{P}
=
\mathbf{I}
$$

we get

$$
\mathrm{tr}(\boldsymbol{\Sigma})
=
\mathrm{tr}
(
\boldsymbol{\Lambda}\mathbf{I}
)
$$

Therefore,

$$
\mathrm{tr}(\boldsymbol{\Sigma})
=
\mathrm{tr}(\boldsymbol{\Lambda})
$$

Since $\boldsymbol{\Lambda}$ is diagonal,

$$
\mathrm{tr}(\boldsymbol{\Lambda})
=
\sum_{j=1}^{m}\lambda_j
$$

Hence,

$$
\mathrm{tr}(\boldsymbol{\Sigma})
=
\mathrm{tr}(\boldsymbol{\Lambda})
=
\sum_{j=1}^{m}\lambda_j
$$

This proves that PCA preserves the total system variance. PCA does not remove variance; it only redistributes the variance into orthogonal principal component directions.

## Cumulative Explained Variance Ratio and Residual Unexplained Variance Ratio

Let the empirical eigenvalues be ordered as

$$
\widehat{\lambda}_1
\geq
\widehat{\lambda}_2
\geq
\cdots
\geq
\widehat{\lambda}_m
$$

The Cumulative Explained Variance Ratio is

$$
\Phi(k)
=
\frac{
\sum_{j=1}^{k}\widehat{\lambda}_j
}{
\sum_{j=1}^{m}\widehat{\lambda}_j
}
$$

The Residual Unexplained Variance Ratio is

$$
\Psi(k)
=
\frac{
\sum_{j=k+1}^{m}\widehat{\lambda}_j
}{
\sum_{j=1}^{m}\widehat{\lambda}_j
}
$$

Since the total variance is divided into explained and unexplained parts,

$$
\Phi(k)+\Psi(k)=1
$$

Therefore,

$$
\Psi(k)=1-\Phi(k)
$$

## 3. Reconstruction, Residual Error, and Energy Decomposition

The centered observation vector is partitioned as

$$
\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n
=
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
+
\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}
$$

The reconstructed vector using the first $k$ principal components is

$$
\widehat{\mathbf{x}}_i
=
\widehat{\boldsymbol{\mu}}_n
+
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
$$

The residual error vector is

$$
\mathbf{e}_i
=
\mathbf{x}_i
-
\widehat{\mathbf{x}}_i
$$

Substituting $\widehat{\mathbf{x}}_i$ gives

$$
\mathbf{e}_i
=
\mathbf{x}_i
-
(
\widehat{\boldsymbol{\mu}}_n
+
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
)
$$

Therefore,

$$
\mathbf{e}_i
=
(
\mathbf{x}_i-\widehat{\boldsymbol{\mu}}_n
)
-
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
$$

Using the PCA partition,

$$
\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n
=
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
+
\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}
$$

we get

$$
\mathbf{e}_i
=
\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}
$$

Now consider the squared Euclidean norm:

$$
\|
\mathbf{x}_i-\widehat{\boldsymbol{\mu}}_n
\|^2
=
\|
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
+
\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}
\|^2
$$

Expanding,

$$
=
\|
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
\|^2
+
\|
\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}
\|^2
+
2
(
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
)^T
(
\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}
)
$$

Since the retained and residual PCA subspaces are orthogonal,

$$
\widehat{\mathbf{P}}_k^T
\widehat{\mathbf{P}}_{m-k}
=
\mathbf{0}
$$

the cross term becomes zero.

Also,

$$
\widehat{\mathbf{P}}_k^T\widehat{\mathbf{P}}_k
=
\mathbf{I}
$$

and

$$
\widehat{\mathbf{P}}_{m-k}^T
\widehat{\mathbf{P}}_{m-k}
=
\mathbf{I}
$$

Therefore,

$$
\|
\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}
\|^2
=
\|
\mathbf{z}_{i,k}
\|^2
$$

and

$$
\|
\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}
\|^2
=
\|
\mathbf{z}_{i,m-k}
\|^2
$$

Hence,

$$
\|
\mathbf{x}_i-\widehat{\boldsymbol{\mu}}_n
\|^2
=
\|
\mathbf{z}_{i,k}
\|^2
+
\|
\mathbf{z}_{i,m-k}
\|^2
$$

This proves that the total squared energy of the centered observation is separated into retained subspace energy and residual subspace energy.

## Hotelling's T-Squared Statistic and Residual Q Statistic

Hotelling's $T^2$ statistic measures abnormal variation inside the retained PCA subspace.

$$
T_i^2
=
\sum_{j=1}^{k}
\frac{z_{i,j}^2}{\widehat{\lambda}_j}
$$

It is sensitive to large but structured changes that still follow the main covariance pattern of the baseline system.

The residual $Q$ statistic, also called SPE, measures the energy outside the retained PCA subspace.

$$
Q_i
=
\sum_{j=k+1}^{m}
z_{i,j}^2
$$

It is sensitive to changes that are not explained by the normal PCA structure.

For an environmental load anomaly such as extreme operational wind loads, Hotelling's $T^2$ is usually more sensitive. This is because the system may still follow the normal structural pattern, but with unusually large variation inside the retained principal component subspace.

For an internal structural fracture such as a fatigue crack, the residual $Q$ statistic is usually more sensitive. This is because a fracture can change the structural load pathway and create a new variation pattern that is not captured by the retained PCA subspace.

Therefore,

$$
\text{Environmental load anomaly} \rightarrow T^2
$$

$$
\text{Internal structural fracture} \rightarrow Q
$$

````python
import numpy as np
import pandas as pd


class PCADiagnosticOptimizer:
    def __init__(self):
        pass

    def analyze_t2_q(self, data, k_values=None):
        """
        Computes PCA-based Hotelling's T^2 and residual Q statistics
        for different truncation boundaries k.

        Parameters
        ----------
        data : pandas.DataFrame or numpy.ndarray
            Baseline data matrix with shape (n, m).
        k_values : list or None
            List of PCA cutoff values. If None, uses k = 1, 2, ..., m-1.

        Returns
        -------
        dict
            PCA eigenvalues, eigenvectors, PCA scores, and summary metrics.
        """

        # Convert input into numeric matrix
        if isinstance(data, pd.DataFrame):
            X = data.select_dtypes(include=[np.number]).to_numpy(dtype=float)
            columns = data.select_dtypes(include=[np.number]).columns.tolist()
        else:
            X = np.asarray(data, dtype=float)
            columns = [f"Sensor_{j+1}" for j in range(X.shape[1])]

        n, m = X.shape

        if n <= m:
            raise ValueError("Number of observations n must be greater than number of sensors m.")

        # Default k values: 1 to m-1
        if k_values is None:
            k_values = list(range(1, m))

        # Validate k values
        for k in k_values:
            if k <= 0 or k >= m:
                raise ValueError("Each k must satisfy 1 <= k < m.")

        # Step 1: Standardize and center the data
        mean_vector = np.mean(X, axis=0)
        std_vector = np.std(X, axis=0, ddof=1)

        # Guardrail for zero-variance sensors
        std_vector[std_vector == 0] = 1e-15

        X_standardized = (X - mean_vector) / std_vector

        # Step 2: Compute unbiased sample covariance matrix
        S = (X_standardized.T @ X_standardized) / (n - 1)

        # Step 3: Eigen decomposition using np.linalg.eigh
        eigenvalues, eigenvectors = np.linalg.eigh(S)

        # Step 4: Sort eigenvalues and eigenvectors in descending order
        sorted_index = np.argsort(eigenvalues)[::-1]
        eigenvalues = eigenvalues[sorted_index]
        eigenvectors = eigenvectors[:, sorted_index]

        # Step 5: Project standardized data into PCA coordinate space
        scores = X_standardized @ eigenvectors

        # Step 6: Compute T^2 and Q for each k
        result_rows = []

        for k in k_values:
            retained_scores = scores[:, :k]
            residual_scores = scores[:, k:]

            retained_eigenvalues = eigenvalues[:k]

            # Avoid division by zero for very small eigenvalues
            retained_eigenvalues = np.maximum(retained_eigenvalues, 1e-15)

            T2_values = np.sum((retained_scores ** 2) / retained_eigenvalues, axis=1)
            Q_values = np.sum(residual_scores ** 2, axis=1)

            result_rows.append({
                "k": k,
                "Mean_T2": np.mean(T2_values),
                "Mean_Q": np.mean(Q_values),
                "Explained_Variance_Ratio": np.sum(eigenvalues[:k]) / np.sum(eigenvalues),
                "Residual_Variance_Ratio": np.sum(eigenvalues[k:]) / np.sum(eigenvalues)
            })

        summary_df = pd.DataFrame(result_rows)

        return {
            "mean_vector": mean_vector,
            "std_vector": std_vector,
            "covariance_matrix": S,
            "eigenvalues": eigenvalues,
            "eigenvectors": eigenvectors,
            "scores": scores,
            "summary": summary_df
        }


# Example usage:
# Replace df_baseline with your real baseline dataset.
# optimizer = PCADiagnosticOptimizer()
# results = optimizer.analyze_t2_q(df_baseline, k_values=[1, 2, 3])
# print("Eigenvalues:")
# print(results["eigenvalues"])
# print("\nT2 and Q Summary:")
# display(results["summary"])
````

Communality represents the amount of sensor variance explained by the common latent factors. In an asset diagnostic system, high communality means the sensor is strongly connected to the shared structural behavior.

## 3. PCA Loading Vector vs Varimax-Rotated FA Loading Vector

A raw PCA loading vector is obtained by rotating the original data space to maximize total variance. PCA focuses on total variance and produces orthogonal principal components ordered by eigenvalues.

A Varimax-rotated FA loading vector is obtained after estimating the latent factor loading matrix and rotating it to improve interpretability. Factor Analysis focuses on shared covariance among variables, not total variance.

Therefore, PCA loadings mainly describe directions of maximum variance, while Varimax-rotated FA loadings describe interpretable latent factors that explain common structure among sensors.

````python
# ============================================================
# Latent Subspace Decomposition via Factor Analysis (FA)
# Full runnable Google Colab code
# ============================================================

import numpy as np
import pandas as pd

from sklearn.decomposition import FactorAnalysis

import plotly.graph_objects as go
from plotly.subplots import make_subplots

from IPython.display import display


# ============================================================
# 1. SAMPLE SENSOR DATA GENERATOR
#    You can replace this df_sensor with your own dataframe later.
# ============================================================

np.random.seed(42)

n_samples = 3000
n_sensors = 6

# Hidden latent factors
latent_factors = np.random.normal(size=(n_samples, 2))

# Loading structure for 6 sensors and 2 hidden factors
true_loadings = np.array([
    [0.85, 0.10],
    [0.80, 0.15],
    [0.75, 0.20],
    [0.10, 0.85],
    [0.15, 0.80],
    [0.20, 0.75]
])

# Sensor-specific noise
noise = np.random.normal(scale=0.35, size=(n_samples, n_sensors))

# Observed sensor data
X = latent_factors @ true_loadings.T + noise

df_sensor = pd.DataFrame(
    X,
    columns=[
        "Sensor_1",
        "Sensor_2",
        "Sensor_3",
        "Sensor_4",
        "Sensor_5",
        "Sensor_6"
    ]
)

display(df_sensor.head())


# ============================================================
# 2. FACTOR ANALYSIS ENGINE
# ============================================================

class FactorAnalysisEngine:
    def __init__(self, df=None):
        self.df = df

    def compute_empirical_fa(self, k: int, columns=None, show_plot: bool = True):
        """
        Factor Analysis latent subspace engine with Varimax rotation.

        This method:
        1. Selects numeric columns.
        2. Standardizes data using ddof=1.
        3. Applies zero-variance protection.
        4. Validates n > m and 1 <= k < m.
        5. Fits Factor Analysis with Varimax rotation.
        6. Computes communality and uniqueness.
        7. Computes Thomson regression factor scores.
        8. Displays a 2 x 2 diagnostic dashboard.
        """

        if self.df is None:
            raise ValueError("No dataframe is available. Please pass a dataframe when creating the class.")

        # Select numeric columns
        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
        else:
            target_cols = list(columns)

        if len(target_cols) == 0:
            raise ValueError("No numeric columns found for Factor Analysis.")

        data = self.df[target_cols].copy().dropna()
        X = data.to_numpy(dtype=float)

        n, m = X.shape

        # Validation rules
        if n <= m:
            raise ValueError("The number of observations n must be greater than the number of features m.")

        if k < 1 or k >= m:
            raise ValueError("The number of latent factors k must satisfy 1 <= k < m.")

        # ====================================================
        # Step 1: Z-score standardization using ddof=1
        # ====================================================

        mean_vector = np.mean(X, axis=0)
        std_vector = np.std(X, axis=0, ddof=1)

        # Zero-variance guardrail
        std_vector[std_vector == 0] = 1e-15

        Z = (X - mean_vector) / std_vector

        # Correlation matrix
        R_matrix = np.corrcoef(Z, rowvar=False)

        # ====================================================
        # Step 2: Fit Factor Analysis with Varimax rotation
        # ====================================================

        fa = FactorAnalysis(
            n_components=k,
            rotation="varimax",
            random_state=42
        )

        fa.fit(Z)

        # Rotated loading matrix: shape = (m, k)
        lambda_rotated = fa.components_.T

        # Uniqueness values
        uniqueness = fa.noise_variance_

        # Communality values
        communality = np.sum(lambda_rotated ** 2, axis=1)

        # ====================================================
        # Step 3: Thomson Regression Factor Scores
        # Formula: F = Z R^{-1} Lambda_rotated
        # ====================================================

        R_inverse = np.linalg.pinv(R_matrix)

        factor_scores = Z @ R_inverse @ lambda_rotated

        # Empirical variance of factor scores
        factor_score_variance = np.var(factor_scores, axis=0, ddof=1)

        # ====================================================
        # Step 4: Output Tables
        # ====================================================

        summary_df = pd.DataFrame({
            "Sensor": target_cols,
            "Communality_h2": communality,
            "Uniqueness_phi2": uniqueness,
            "Communality_%": communality * 100,
            "Uniqueness_%": uniqueness * 100
        })

        factor_variance_df = pd.DataFrame({
            "Factor": [f"Factor {i + 1}" for i in range(k)],
            "Empirical_Variance": factor_score_variance
        })

        loading_df = pd.DataFrame(
            lambda_rotated,
            index=target_cols,
            columns=[f"Factor {i + 1}" for i in range(k)]
        )

        print("\n===================================================")
        print("Factor Analysis Latent Subspace Decomposition")
        print("===================================================")
        print(f"Number of observations (n): {n}")
        print(f"Number of sensors/features (m): {m}")
        print(f"Number of latent factors (k): {k}")
        print(f"Average System Communality: {np.mean(communality) * 100:.2f}%")
        print(f"Average System Uniqueness: {np.mean(uniqueness) * 100:.2f}%")

        print("\nRotated Loading Matrix:")
        display(loading_df)

        print("\nCommunality and Uniqueness Summary:")
        display(summary_df)

        print("\nLatent Factor Score Variance:")
        display(factor_variance_df)

        # ====================================================
        # Step 5: Plotly 2 x 2 Dashboard
        # ====================================================

        if show_plot:
            sensor_labels = target_cols
            factor_labels = [f"Factor {i + 1}" for i in range(k)]

            fig = make_subplots(
                rows=2,
                cols=2,
                horizontal_spacing=0.24,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings Matrix Heatmap",
                    "Variance Partitioning Profile",
                    "Sensor Uniqueness Line Profile",
                    "Latent Factor Variance Distribution"
                )
            )

            # ------------------------------------------------
            # Subplot (1,1): Structural Loadings Matrix Heatmap
            # ------------------------------------------------

            fig.add_trace(
                go.Heatmap(
                    z=np.abs(lambda_rotated),
                    x=factor_labels,
                    y=sensor_labels,
                    colorscale="YlOrRd",
                    colorbar=dict(
                        title="|Λ rotated|",
                        x=-0.15,
                        len=0.38,
                        y=0.78,
                        yanchor="middle",
                        xanchor="right"
                    )
                ),
                row=1,
                col=1
            )

            # ------------------------------------------------
            # Subplot (1,2): Variance Partitioning Profile
            # ------------------------------------------------

            fig.add_trace(
                go.Bar(
                    y=sensor_labels,
                    x=communality * 100,
                    name="Shared Communality (h²)",
                    orientation="h",
                    marker=dict(color="#1f77b4")
                ),
                row=1,
                col=2
            )

            fig.add_trace(
                go.Bar(
                    y=sensor_labels,
                    x=uniqueness * 100,
                    name="Isolated Uniqueness (φ²)",
                    orientation="h",
                    marker=dict(color="#ff7f0e")
                ),
                row=1,
                col=2
            )

            # ------------------------------------------------
            # Subplot (2,1): Sensor Uniqueness Line Profile
            # ------------------------------------------------

            fig.add_trace(
                go.Scatter(
                    x=sensor_labels,
                    y=uniqueness,
                    mode="lines+markers",
                    name="Sensor Uniqueness (φ²)",
                    line=dict(
                        color="#d62728",
                        width=2,
                        dash="dashdot"
                    ),
                    marker=dict(
                        size=8,
                        symbol="x"
                    )
                ),
                row=2,
                col=1
            )

            # ------------------------------------------------
            # Subplot (2,2): Latent Factor Variance Distribution
            # ------------------------------------------------

            fig.add_trace(
                go.Bar(
                    x=factor_labels,
                    y=factor_score_variance,
                    name="Latent Factor Variance",
                    marker=dict(
                        color="#2ca02c",
                        line=dict(
                            color="black",
                            width=0.5
                        )
                    )
                ),
                row=2,
                col=2
            )

            # Axis labels
            fig.update_xaxes(title_text="Latent Factors", row=1, col=1)
            fig.update_yaxes(title_text="Sensors", row=1, col=1)

            fig.update_xaxes(
                title_text="Variance Percentage (%)",
                range=[0, 100],
                row=1,
                col=2
            )
            fig.update_yaxes(title_text="Sensors", row=1, col=2)

            fig.update_xaxes(
                title_text="Sensors",
                tickangle=25,
                row=2,
                col=1
            )
            fig.update_yaxes(title_text="Uniqueness φ²", row=2, col=1)

            fig.update_xaxes(title_text="Latent Factors", row=2, col=2)
            fig.update_yaxes(title_text="Empirical Variance", row=2, col=2)

            # Layout settings
            fig.update_layout(
                title=dict(
                    text="2 × 2 Factor Analysis Diagnostic Dashboard",
                    x=0.5,
                    y=0.97,
                    xanchor="center",
                    yanchor="top"
                ),
                template="plotly_white",
                barmode="stack",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    x=0.5,
                    y=1.02,
                    xanchor="center"
                ),
                margin=dict(t=150, b=60, l=140, r=80),
                width=1250,
                height=750
            )

            fig.show()

        return {
            "mean_vector": mean_vector,
            "std_vector": std_vector,
            "standardized_Z": Z,
            "correlation_matrix_R": R_matrix,
            "loading_matrix_Lambda_rotated": lambda_rotated,
            "uniqueness_vector_phi2": uniqueness,
            "communality_vector_h2": communality,
            "factor_scores_Thomson": factor_scores,
            "factor_score_variance": factor_score_variance,
            "loading_dataframe": loading_df,
            "summary_dataframe": summary_df,
            "factor_variance_dataframe": factor_variance_df
        }


# ============================================================
# 3. RUN THE FACTOR ANALYSIS ENGINE
# ============================================================

engine = FactorAnalysisEngine(df_sensor)

results = engine.compute_empirical_fa(
    k=2,
    columns=[
        "Sensor_1",
        "Sensor_2",
        "Sensor_3",
        "Sensor_4",
        "Sensor_5",
        "Sensor_6"
    ],
    show_plot=True
)
````

````python
import numpy as np
import pandas as pd

from typing import Optional, Sequence, Dict, Any
from sklearn.decomposition import FactorAnalysis

import plotly.graph_objects as go
from plotly.subplots import make_subplots

from IPython.display import display


# ============================================================
# Synthetic 4-Sensor Dataset
# ============================================================

np.random.seed(42)

n_samples = 3000

# Hidden physical processes
f1 = np.random.normal(0, 1, n_samples)
f2 = np.random.normal(0, 1, n_samples)

# Normal sensor noise
e1 = np.random.normal(0, 0.20, n_samples)
e2 = np.random.normal(0, 0.20, n_samples)
e3 = np.random.normal(0, 0.25, n_samples)

# High isolated noise for Sensor_4
e4 = np.random.normal(0, np.sqrt(2.0), n_samples)

sensor_1 = 1.00 * f1 + 0.10 * f2 + e1
sensor_2 = 0.90 * f1 + 0.15 * f2 + e2
sensor_3 = 0.10 * f1 + 1.00 * f2 + e3
sensor_4 = e4

df_subspace = pd.DataFrame({
    "Sensor_1": sensor_1,
    "Sensor_2": sensor_2,
    "Sensor_3": sensor_3,
    "Sensor_4": sensor_4
})

display(df_subspace.head())


# ============================================================
# Dual PCA and FA Diagnostic Engine
# ============================================================

class DualSubspaceDiagnosticsEngine:
    def __init__(self, df: pd.DataFrame):
        self.df = df

    def compute_empirical_pca(
        self,
        columns: Optional[Sequence[str]] = None,
        show_plot: bool = True
    ) -> Dict[str, Any]:

        if self.df is None:
            raise ValueError("No dataframe is available.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().to_numpy(dtype=float)

        n, m = X.shape

        if n <= m:
            raise ValueError("Sample size n must be greater than feature count m.")

        # Center data
        mean_vector = np.mean(X, axis=0)
        X_centered = X - mean_vector

        # Unbiased covariance matrix
        S = np.cov(X_centered, rowvar=False, ddof=1)

        # Eigen decomposition
        eigenvalues, eigenvectors = np.linalg.eigh(S)

        # Sort descending
        sort_index = np.argsort(eigenvalues)[::-1]
        eigenvalues = eigenvalues[sort_index]
        eigenvectors = eigenvectors[:, sort_index]

        eigenvalues = np.clip(eigenvalues, 1e-15, None)

        # Scores
        scores = X_centered @ eigenvectors

        total_variance = np.sum(eigenvalues)
        explained_ratio = eigenvalues / total_variance
        cumulative_ratio = np.cumsum(explained_ratio)
        residual_ratio = 1 - cumulative_ratio

        # Mean T2 and Q for k = 1, 2, 3
        k_values = list(range(1, m))

        mean_t2 = []
        mean_q = []

        for k in k_values:
            retained_scores = scores[:, :k]
            retained_eigenvalues = eigenvalues[:k]

            t2_values = np.sum((retained_scores ** 2) / retained_eigenvalues, axis=1)
            q_values = np.sum(scores[:, k:] ** 2, axis=1)

            mean_t2.append(np.mean(t2_values))
            mean_q.append(np.mean(q_values))

        loading_df = pd.DataFrame(
            eigenvectors,
            index=target_cols,
            columns=[f"PC {i+1}" for i in range(m)]
        )

        variance_df = pd.DataFrame({
            "Component": [f"PC {i+1}" for i in range(m)],
            "Eigenvalue": eigenvalues,
            "Explained_%": explained_ratio * 100,
            "Cumulative_%": cumulative_ratio * 100,
            "Residual_%": residual_ratio * 100
        })

        tq_df = pd.DataFrame({
            "k": k_values,
            "Mean_T2": mean_t2,
            "Mean_Q": mean_q
        })

        print("\nPCA Loading Matrix")
        display(loading_df)

        print("\nPCA Variance Summary")
        display(variance_df)

        print("\nMean T2 and Q Summary")
        display(tq_df)

        if show_plot:
            pc_labels = [f"PC {i+1}" for i in range(m)]
            k_labels = [f"k={k}" for k in k_values]

            fig = make_subplots(
                rows=2,
                cols=3,
                horizontal_spacing=0.16,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Feature Loadings Matrix |Pj,r|",
                    "Absolute Eigenvalues",
                    "Explained and Cumulative Variance",
                    "Residual Space",
                    "Mean Hotelling's T² vs k",
                    "Mean Q Statistic (SPE) vs k"
                )
            )

            # Panel 1,1
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(eigenvectors),
                    x=pc_labels,
                    y=target_cols,
                    colorscale="YlOrRd",
                    colorbar=dict(
                        title="|Pj,r|",
                        x=-0.12,
                        len=0.38,
                        y=0.78,
                        yanchor="middle",
                        xanchor="right"
                    )
                ),
                row=1,
                col=1
            )

            # Panel 1,2
            fig.add_trace(
                go.Bar(
                    x=pc_labels,
                    y=eigenvalues,
                    name="Eigenvalue",
                    marker=dict(color="#1f77b4")
                ),
                row=1,
                col=2
            )

            # Panel 1,3
            fig.add_trace(
                go.Bar(
                    x=pc_labels,
                    y=explained_ratio * 100,
                    name="Explained Variance %",
                    marker=dict(color="#ff7f0e")
                ),
                row=1,
                col=3
            )

            fig.add_trace(
                go.Scatter(
                    x=pc_labels,
                    y=cumulative_ratio * 100,
                    mode="lines+markers",
                    name="Cumulative Variance %",
                    line=dict(color="#2ca02c", dash="dash", width=2)
                ),
                row=1,
                col=3
            )

            # Panel 2,1
            fig.add_trace(
                go.Bar(
                    x=pc_labels,
                    y=residual_ratio * 100,
                    name="Residual Variance %",
                    marker=dict(color="#d62728")
                ),
                row=2,
                col=1
            )

            # Panel 2,2
            fig.add_trace(
                go.Scatter(
                    x=k_labels,
                    y=mean_t2,
                    mode="lines+markers",
                    name="Mean T²",
                    line=dict(color="#9467bd", width=2)
                ),
                row=2,
                col=2
            )

            # Panel 2,3
            fig.add_trace(
                go.Scatter(
                    x=k_labels,
                    y=mean_q,
                    mode="lines+markers",
                    name="Mean Q",
                    line=dict(color="#2ca02c", width=2)
                ),
                row=2,
                col=3
            )

            fig.update_xaxes(title_text="Principal Components", row=1, col=1)
            fig.update_yaxes(title_text="Sensors", row=1, col=1)

            fig.update_xaxes(title_text="Principal Components", row=1, col=2)
            fig.update_yaxes(title_text="Eigenvalue", row=1, col=2)

            fig.update_xaxes(title_text="Principal Components", row=1, col=3)
            fig.update_yaxes(title_text="Variance %", range=[0, 100], row=1, col=3)

            fig.update_xaxes(title_text="Principal Components", row=2, col=1)
            fig.update_yaxes(title_text="Residual Variance %", range=[0, 100], row=2, col=1)

            fig.update_xaxes(title_text="Subspace Size k", row=2, col=2)
            fig.update_yaxes(title_text="Mean T²", row=2, col=2)

            fig.update_xaxes(title_text="Subspace Size k", row=2, col=3)
            fig.update_yaxes(title_text="Mean Q", row=2, col=3)

            fig.update_layout(
                title=dict(
                    text="PCA Optimization Suite",
                    x=0.5,
                    y=0.97,
                    xanchor="center",
                    yanchor="top"
                ),
                template="plotly_white",
                width=1450,
                height=800,
                showlegend=True,
                legend=dict(
                    orientation="h",
                    x=0.5,
                    y=1.02,
                    xanchor="center"
                ),
                margin=dict(t=150, b=60, l=120, r=80)
            )

            fig.show()

        return {
            "loading_df": loading_df,
            "variance_df": variance_df,
            "tq_df": tq_df,
            "eigenvalues": eigenvalues,
            "eigenvectors": eigenvectors,
            "scores": scores
        }

    def compute_empirical_fa(
        self,
        k: int,
        columns: Optional[Sequence[str]] = None,
        show_plot: bool = True
    ) -> Dict[str, Any]:

        if self.df is None:
            raise ValueError("No dataframe is available.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().to_numpy(dtype=float)

        n, m = X.shape

        if n <= m:
            raise ValueError("Sample size n must be greater than feature count m.")

        if k < 1 or k >= m:
            raise ValueError("k must satisfy 1 <= k < m.")

        # Z-score standardization
        mean_vector = np.mean(X, axis=0)
        std_vector = np.std(X, axis=0, ddof=1)

        std_vector[std_vector == 0] = 1e-15

        Z = (X - mean_vector) / std_vector

        R = np.corrcoef(Z, rowvar=False)

        # Varimax Factor Analysis
        fa = FactorAnalysis(
            n_components=k,
            rotation="varimax",
            random_state=42
        )

        fa.fit(Z)

        lambda_rotated = fa.components_.T

        communality = np.sum(lambda_rotated ** 2, axis=1)
        communality = np.clip(communality, 0, 1)

        uniqueness = 1 - communality
        uniqueness = np.clip(uniqueness, 0, 1)

        # Thomson scores
        R_inverse = np.linalg.pinv(R)
        factor_scores = Z @ R_inverse @ lambda_rotated

        factor_variance = np.var(factor_scores, axis=0, ddof=1)

        loading_df = pd.DataFrame(
            lambda_rotated,
            index=target_cols,
            columns=[f"Factor {i+1}" for i in range(k)]
        )

        fa_summary = pd.DataFrame({
            "Sensor": target_cols,
            "Communality_h2": communality,
            "Uniqueness_phi2": uniqueness,
            "Communality_%": communality * 100,
            "Uniqueness_%": uniqueness * 100
        })

        factor_variance_df = pd.DataFrame({
            "Factor": [f"Factor {i+1}" for i in range(k)],
            "Empirical_Variance": factor_variance
        })

        print("\nRotated FA Loading Matrix")
        display(loading_df)

        print("\nFA Communality and Uniqueness Summary")
        display(fa_summary)

        print("\nLatent Factor Variance")
        display(factor_variance_df)

        if show_plot:
            factor_labels = [f"Factor {i+1}" for i in range(k)]

            fig = make_subplots(
                rows=2,
                cols=2,
                horizontal_spacing=0.24,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings Matrix Heatmap",
                    "Variance Partitioning Profile",
                    "Sensor Uniqueness Line Profile",
                    "Latent Factor Variance Distribution"
                )
            )

            # Subplot 1,1
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(lambda_rotated),
                    x=factor_labels,
                    y=target_cols,
                    colorscale="YlOrRd",
                    colorbar=dict(
                        title="|Λ rotated|",
                        x=-0.15,
                        len=0.38,
                        y=0.78,
                        yanchor="middle",
                        xanchor="right"
                    )
                ),
                row=1,
                col=1
            )

            # Subplot 1,2
            fig.add_trace(
                go.Bar(
                    y=target_cols,
                    x=communality * 100,
                    name="Shared Communality h²",
                    orientation="h",
                    marker=dict(color="#1f77b4")
                ),
                row=1,
                col=2
            )

            fig.add_trace(
                go.Bar(
                    y=target_cols,
                    x=uniqueness * 100,
                    name="Channel Uniqueness φ²",
                    orientation="h",
                    marker=dict(color="#ff7f0e")
                ),
                row=1,
                col=2
            )

            # Subplot 2,1
            fig.add_trace(
                go.Scatter(
                    x=target_cols,
                    y=uniqueness,
                    mode="lines+markers",
                    name="Uniqueness φ²",
                    line=dict(
                        color="#d62728",
                        width=2,
                        dash="dashdot"
                    ),
                    marker=dict(
                        size=8,
                        symbol="x"
                    )
                ),
                row=2,
                col=1
            )

            # Subplot 2,2
            fig.add_trace(
                go.Bar(
                    x=factor_labels,
                    y=factor_variance,
                    name="Factor Variance",
                    marker=dict(
                        color="#2ca02c",
                        line=dict(
                            color="black",
                            width=0.5
                        )
                    )
                ),
                row=2,
                col=2
            )

            fig.update_xaxes(title_text="Latent Factors", row=1, col=1)
            fig.update_yaxes(title_text="Sensors", row=1, col=1)

            fig.update_xaxes(
                title_text="Variance Percentage (%)",
                range=[0, 100],
                row=1,
                col=2
            )
            fig.update_yaxes(title_text="Sensors", row=1, col=2)

            fig.update_xaxes(
                title_text="Sensors",
                tickangle=25,
                row=2,
                col=1
            )
            fig.update_yaxes(title_text="Uniqueness φ²", row=2, col=1)

            fig.update_xaxes(title_text="Latent Factors", row=2, col=2)
            fig.update_yaxes(title_text="Empirical Variance", row=2, col=2)

            fig.update_layout(
                title=dict(
                    text="FA Latent Subspace Suite",
                    x=0.5,
                    y=0.97,
                    xanchor="center",
                    yanchor="top"
                ),
                template="plotly_white",
                barmode="stack",
                width=1250,
                height=750,
                showlegend=True,
                legend=dict(
                    orientation="h",
                    x=0.5,
                    y=1.02,
                    xanchor="center"
                ),
                margin=dict(t=150, b=60, l=140, r=80)
            )

            fig.show()

        return {
            "loading_df": loading_df,
            "fa_summary": fa_summary,
            "factor_variance_df": factor_variance_df,
            "communality": communality,
            "uniqueness": uniqueness,
            "factor_scores": factor_scores
        }


# ============================================================
# Run the dashboards
# ============================================================

columns = ["Sensor_1", "Sensor_2", "Sensor_3", "Sensor_4"]

engine = DualSubspaceDiagnosticsEngine(df_subspace)

pca_results = engine.compute_empirical_pca(
    columns=columns,
    show_plot=True
)

fa_results = engine.compute_empirical_fa(
    k=2,
    columns=columns,
    show_plot=True
)
````

## Question 1: PCA Loading Contamination by Sensor Noise

Sensor_4 is different from the other sensors because it is not connected to the true hidden physical processes of the system.

Sensor_1, Sensor_2, and Sensor_3 are related to the real structural behavior, but Sensor_4 mainly contains isolated noise or fault energy.

In the PCA loading matrix, PC 1 becomes strongly dominated by Sensor_4. This happens because PCA always captures the direction with the largest total variance first.

Since Sensor_4 has very large noise variance, PCA treats it as the most important direction. Therefore, PC 1 mainly tracks Sensor_4 instead of the real shared structural behavior.

This creates a monitoring risk. A single noisy or faulty sensor can control the main PCA axis and hide the true behavior of the asset.

So, standard PCA can become unreliable when one sensor has very high isolated noise.

## Question 2: Decoupling Structural Loading via Rotation

PCA follows strict variance maximization. Each principal component tries to capture as much total variance as possible.

Because of this, PCA components can become mixed. One component may contain the effects of several sensors at the same time.

This makes the PCA loading matrix harder to interpret. If an alarm appears on one PCA component, the engineer must identify which sensor or physical process caused it.

Factor Analysis uses Varimax rotation to create a simpler structure. Varimax rotation makes each factor strongly connected to a smaller and clearer group of sensors.

After rotation, Sensor_1 and Sensor_2 can group clearly under one factor, while Sensor_3 can move mainly into another factor. Sensor_4 can remain separated as a high-uniqueness sensor.

Therefore, the rotated FA heatmap is easier to troubleshoot. It gives cleaner sensor groupings and helps the operator identify the root cause faster than the raw PCA loading matrix.

## Question 3: Determining Subspace Truncation \(k\)

The Mean Q Statistic shows how much residual information remains outside the selected PCA subspace.

When the cutoff changes from

$$
k = 1
$$

to

$$
k = 2
$$

the Mean Q Statistic drops sharply. This means that the second component captures a large amount of important remaining structure.

When the cutoff changes from

$$
k = 2
$$

to

$$
k = 3
$$

the Q curve becomes flatter. This creates an elbow pattern.

A sharp drop followed by a flat elbow shows that

$$
k = 2
$$

is the best hidden physical dimensionality of the asset.

This means two main dimensions are enough to explain the important structure.

If

$$
k = 3
$$

is selected, residual noise is forced into the clean principal subspace. This can overfit the monitoring model and make the system react to noise instead of meaningful structural behavior.

Therefore, the better truncation choice is

$$
k = 2
$$

## Question 4: Operational Trade-offs in System Health Monitoring

The PCA strategy monitors two spaces.

The first space is the retained principal subspace using Hotelling's

$$
T^2
$$

statistic.

The second space is the residual space using the

$$
Q
$$

statistic.

This approach is useful because

$$
T^2
$$

tracks abnormal behavior inside the main PCA subspace, while

$$
Q
$$

tracks abnormal residual variation outside the selected subspace.

However, PCA can be affected by one faulty sensor. If one sensor loses calibration or has an electrical short, its large variance can rotate the PCA axes and dominate the monitoring model.

The FA strategy monitors rotated latent factor scores and separates shared structural variance from sensor-specific uniqueness.

If one sensor becomes faulty, its uniqueness value

$$
\phi_j^2
$$

will increase.

This helps identify that the issue is local to that sensor rather than a full structural failure.

Therefore, Factor Analysis is more robust against a single sensor losing calibration or experiencing an electrical short.

This is because FA can place the faulty sensor behavior into the uniqueness noise floor while keeping the shared latent factor structure more stable.
