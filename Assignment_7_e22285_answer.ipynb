# Q. Analytical Derivation

Consider the filter model

$$
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
$$

where

$$
x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),
$$

Answer the followings:

1. Show that

$$
x_k^- \sim \mathscr{N}(m_k^-, P_k^-),
$$

where

$$
m_k^- = A_{k-1}m_{k-1},
$$

and

$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T
+ G_{k-1}\Sigma_p G_{k-1}^T.
$$

2. Show that

$$
y_k^- \sim
\mathscr{N}
\left(
H_k m_k^-,
H_k P_k^- H_k^T + \Sigma_m
\right).
$$

3. Show that

$$
\begin{bmatrix}
x_k^- \\
y^{-}_k
\end{bmatrix}
\sim
\mathscr{N}
\left(
\begin{bmatrix}
m_k^- \\
H_k m_k^-
\end{bmatrix},
\begin{bmatrix}
P_k^- & P_k^- H_k^T \\
H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m
\end{bmatrix}
\right).
$$

4. Show that

$$
x^{+}_k
\triangleq
\left(
x^{-}_k
\mid
y^{-}_k = y^{\mathrm{obs}}_{k}
\right)
\sim
\mathscr{N}(m_k, P_k),
$$

where the updated mean $m_k$ and updated covariance $P_k$ are given by:

$$
\begin{align*}
K_k
&\triangleq
P_k^- H_k^T
\left(
H_k P_k^- H_k^T + \Sigma_m
\right)^{-1}, \\
m_k
&=
m_k^- +
K_k
\left(
y^{\mathrm{obs}}_{k} - H_k m_k^-
\right), \\
P_k
&=
(I - K_k H_k) P_k^-.
\end{align*}
$$

5. Find

$$
\mathbb{E}
\left[
x^{-}_k
\mid
y^{-}_k = y^{\mathrm{obs}}_{k}
\right]
$$

and

$$
\operatorname{Var}
\left(
x^{-}_k
\mid
y^{-}_k = y^{\mathrm{obs}}_{k}
\right).
$$

## Answer

### 1.

The state equation is a linear combination of independent Gaussian variables. Therefore, \(x_k^-\) is Gaussian.

Its mean is

$$
\begin{aligned}
\mathbb{E}[x_k^-]
&=
\mathbb{E}
\left[
A_{k-1}x_{k-1}^+
+
G_{k-1}w_{k-1}
\right] \\
&=
A_{k-1}\mathbb{E}[x_{k-1}^+]
+
G_{k-1}\mathbb{E}[w_{k-1}] \\
&=
A_{k-1}m_{k-1}.
\end{aligned}
$$

Thus,

$$
m_k^- = A_{k-1}m_{k-1}.
$$

Its covariance is

$$
\begin{aligned}
\operatorname{Var}(x_k^-)
&=
\operatorname{Var}
\left(
A_{k-1}x_{k-1}^+
+
G_{k-1}w_{k-1}
\right) \\
&=
A_{k-1}P_{k-1}A_{k-1}^T
+
G_{k-1}\Sigma_pG_{k-1}^T.
\end{aligned}
$$

Therefore,

$$
x_k^-
\sim
\mathscr{N}(m_k^-,P_k^-).
$$

### 2.

From

$$
y_k^- = H_kx_k^- + z_k,
$$

the mean is

$$
\mathbb{E}[y_k^-]
=
H_km_k^-.
$$

The covariance is

$$
\operatorname{Var}(y_k^-)
=
H_kP_k^-H_k^T+\Sigma_m.
$$

Hence,

$$
y_k^-
\sim
\mathscr{N}
\left(
H_km_k^-,
H_kP_k^-H_k^T+\Sigma_m
\right).
$$

### 3.

The cross-covariance is

$$
\begin{aligned}
\operatorname{Cov}(x_k^-,y_k^-)
&=
\operatorname{Cov}
\left(
x_k^-,
H_kx_k^-+z_k
\right) \\
&=
P_k^-H_k^T.
\end{aligned}
$$

Therefore,

$$
\begin{bmatrix}
x_k^- \\
y_k^-
\end{bmatrix}
\sim
\mathscr{N}
\left(
\begin{bmatrix}
m_k^- \\
H_km_k^-
\end{bmatrix},
\begin{bmatrix}
P_k^- & P_k^-H_k^T \\
H_kP_k^- & H_kP_k^-H_k^T+\Sigma_m
\end{bmatrix}
\right).
$$

### 4.

Using the conditional Gaussian formula,

$$
K_k
=
P_k^-H_k^T
\left(
H_kP_k^-H_k^T+\Sigma_m
\right)^{-1}.
$$

The conditional mean is

$$
m_k
=
m_k^-
+
K_k
\left(
y_k^{\mathrm{obs}}-H_km_k^-
\right).
$$

The conditional covariance is

$$
P_k
=
(I-K_kH_k)P_k^-.
$$

Thus,

$$
x_k^+
\sim
\mathscr{N}(m_k,P_k).
$$

### 5.

$$
\mathbb{E}
\left[
x_k^-
\mid
y_k^-=y_k^{\mathrm{obs}}
\right]
=
m_k,
$$

and

$$
\operatorname{Var}
\left(
x_k^-
\mid
y_k^-=y_k^{\mathrm{obs}}
\right)
=
P_k.
$$

# Q. 1-D Example

Consider the scalar linear-Gaussian filter model

$$
\begin{aligned}
x^-_k
&=
a\,x^+_{k-1}+w_{k-1},
\qquad
w_{k-1}\sim\mathscr{N}(0,q), \\
y^-_k
&=
h\,x^-_k+z_k,
\qquad
z_k\sim\mathscr{N}(0,r),
\end{aligned}
$$

with prior

$$
x_0\sim\mathscr{N}(m_0,P_0).
$$

Define

$$
Y_k
=
\left\{
y^{\mathrm{obs}}_1,
\dots,
y^{\mathrm{obs}}_k
\right\}.
$$

Answer the following:

1. Show that

$$
\begin{aligned}
m_k^- &= a\,m_{k-1}, \\
P_k^- &= a^2P_{k-1}+q.
\end{aligned}
$$

2. Show that

$$
\begin{aligned}
m_k
&=
m_k^-+K_kv_k \\
&=
m_k^-+
\frac{P_k^-h}{S_k}
\left(
y_k^{\mathrm{obs}}-hm_k^-
\right), \\
P_k
&=
(1-K_kh)P_k^- \\
&=
\left(
1-\frac{P_k^-h^2}{S_k}
\right)P_k^-.
\end{aligned}
$$

3. Show that the predictive measurement distribution (before seeing \(y_k\)):

$$
p(y_k^-\mid Y_{k-1})
=
\mathscr{N}
\left(
hm_k^-,
h^2P_k^-+r
\right).
$$

4. Show that the posterior-predictive measurement distribution (after filtering on \(y_k\)):

$$
p(y_k^-\mid Y_k)
=
\mathscr{N}
\left(
hm_k,
h^2P_k+r
\right).
$$

5. For suitable numerical values for \(a,q,h,r\) animate the above to prior distribution and the posterior distribution.

## Answer

### 1.

Using the state equation,

$$
\begin{aligned}
m_k^-
&=
\mathbb{E}
\left[
ax_{k-1}^+ + w_{k-1}
\right] \\
&=
am_{k-1},
\end{aligned}
$$

and

$$
\begin{aligned}
P_k^-
&=
\operatorname{Var}
\left(
ax_{k-1}^+ + w_{k-1}
\right) \\
&=
a^2P_{k-1}+q.
\end{aligned}
$$

### 2.

The innovation and its variance are

$$
v_k
=
y_k^{\mathrm{obs}}-hm_k^-,
$$

and

$$
S_k
=
h^2P_k^-+r.
$$

The Kalman gain is

$$
K_k
=
\frac{P_k^-h}{S_k}.
$$

Therefore,

$$
\begin{aligned}
m_k
&=
m_k^-+K_kv_k \\
&=
m_k^-+
\frac{P_k^-h}{S_k}
\left(
y_k^{\mathrm{obs}}-hm_k^-
\right),
\end{aligned}
$$

and

$$
\begin{aligned}
P_k
&=
(1-K_kh)P_k^- \\
&=
\left(
1-\frac{P_k^-h^2}{S_k}
\right)P_k^-.
\end{aligned}
$$

### 3.

Before receiving \(y_k\),

$$
y_k^-
=
hx_k^-+z_k.
$$

Therefore,

$$
p(y_k^-\mid Y_{k-1})
=
\mathscr{N}
\left(
hm_k^-,
h^2P_k^-+r
\right).
$$

### 4.

After updating the state with \(y_k\),

$$
p(y_k^-\mid Y_k)
=
\mathscr{N}
\left(
hm_k,
h^2P_k+r
\right).
$$

### 5.

The following code animates the change from the prior distribution to the posterior distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

a, q, h, r = 1.0, 0.2, 1.0, 0.5
m, P = 0.0, 1.0
y_obs = 2.0

x = np.linspace(-4, 4, 400)

def pdf(x, mean, variance):
    return (
        1 / np.sqrt(2 * np.pi * variance)
        * np.exp(-(x - mean)**2 / (2 * variance))
    )

# Prior values
m_prior = a * m
P_prior = a**2 * P + q

# Posterior values
S = h**2 * P_prior + r
K = P_prior * h / S

m_post = m_prior + K * (y_obs - h * m_prior)
P_post = (1 - K * h) * P_prior

fig, ax = plt.subplots(figsize=(8, 5))

prior, = ax.plot([], [], label="Prior")
posterior, = ax.plot([], [], label="Posterior")

ax.set_xlim(-4, 4)
ax.set_ylim(0, 1)
ax.set_xlabel("State value")
ax.set_ylabel("Probability density")
ax.set_title("Prior and Posterior Distributions")
ax.grid(True)
ax.legend()

def update(i):
    t = i / 30

    current_mean = (1 - t) * m_prior + t * m_post
    current_variance = (1 - t) * P_prior + t * P_post

    prior.set_data(
        x,
        pdf(x, m_prior, P_prior)
    )

    posterior.set_data(
        x,
        pdf(x, current_mean, current_variance)
    )

    return prior, posterior

ani = FuncAnimation(
    fig,
    update,
    frames=31,
    interval=100,
    blit=True
)

plt.close(fig)

HTML(ani.to_jshtml())
```

# Q. 2D-Position Estimation

## Part A

Consider a two-dimensional constant-velocity dynamical system. The hidden state at time step \(k\) is

$$
x_k =
\begin{bmatrix}
p_x(k) \\
p_y(k) \\
v_x(k) \\
v_y(k)
\end{bmatrix},
$$

where \((p_x(k),p_y(k))\) denote the position components and \((v_x(k),v_y(k))\) denote the velocity components.

The measurement consists only of the two position components:

$$
y_k =
\begin{bmatrix}
p_x^{\mathrm{meas}}(k) \\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}.
$$

The linear Gaussian state-space filter model is

$$
x^-_k = A x^+_{k-1} + G w_{k-1},
$$

$$
y^-_k = Hx^+_k + z_k,
$$

where

$$
w_{k-1} \sim \mathscr{N}(0,\Sigma_p),
\qquad
z_k \sim \mathscr{N}(0,\Sigma_m).
$$

The process noise sequence \(w_k\), measurement noise sequence \(z_k\), and the initial state are assumed mutually independent.

Show that:

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0 \\
0 & 1 & 0 & \Delta t \\
0 & 0 & 1 & 0 \\
0 & 0 & 0 & 1
\end{bmatrix},
$$

$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0 \\
0 & 1 & 0 & 0
\end{bmatrix},
$$

and

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0 \\
0 & \frac{1}{2}\Delta t^2 \\
\Delta t & 0 \\
0 & \Delta t
\end{bmatrix}.
$$

## Answer

Under the constant-velocity assumption,

$$
\begin{aligned}
p_x(k)
&=
p_x(k-1)+\Delta t\,v_x(k-1), \\
p_y(k)
&=
p_y(k-1)+\Delta t\,v_y(k-1), \\
v_x(k)
&=
v_x(k-1), \\
v_y(k)
&=
v_y(k-1).
\end{aligned}
$$

Writing these equations in matrix form gives

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0 \\
0 & 1 & 0 & \Delta t \\
0 & 0 & 1 & 0 \\
0 & 0 & 0 & 1
\end{bmatrix}.
$$

The measurement contains only \(p_x\) and \(p_y\). Therefore,

$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0 \\
0 & 1 & 0 & 0
\end{bmatrix}.
$$

Let the acceleration noise be

$$
w_{k-1}
=
\begin{bmatrix}
a_x \\
a_y
\end{bmatrix}.
$$

For each direction,

$$
\Delta p
=
\frac{1}{2}a\Delta t^2,
\qquad
\Delta v
=
a\Delta t.
$$

Hence,

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0 \\
0 & \frac{1}{2}\Delta t^2 \\
\Delta t & 0 \\
0 & \Delta t
\end{bmatrix}.
$$

## Part B

Develop a python based method to filter a sequence of noisy GPS obtained position meaurements.

## Answer

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate sample GPS data
np.random.seed(10)

dt = 1.0
N = 100

# True trajectory
true_x = np.zeros((4, N))
true_x[:, 0] = [0, 0, 1.5, 0.8]

for k in range(1, N):
    true_x[0, k] = true_x[0, k-1] + dt * true_x[2, k-1]
    true_x[1, k] = true_x[1, k-1] + dt * true_x[3, k-1]
    true_x[2, k] = true_x[2, k-1]
    true_x[3, k] = true_x[3, k-1]

# Noisy GPS measurements
gps_std = 2.5

measurements = np.zeros((2, N))

measurements[0] = (
    true_x[0]
    + np.random.normal(0, gps_std, N)
)

measurements[1] = (
    true_x[1]
    + np.random.normal(0, gps_std, N)
)

# Kalman filter matrices
A = np.array([
    [1, 0, dt, 0],
    [0, 1, 0, dt],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

H = np.array([
    [1, 0, 0, 0],
    [0, 1, 0, 0]
])

G = np.array([
    [0.5 * dt**2, 0],
    [0, 0.5 * dt**2],
    [dt, 0],
    [0, dt]
])

# Noise matrices
sigma_acc = 0.2
Q = sigma_acc**2 * np.eye(2)

R = gps_std**2 * np.eye(2)

# Initial values
x = np.array([0, 0, 0, 0], dtype=float)
P = np.eye(4) * 20

filtered = np.zeros((4, N))

# Kalman filter loop
for k in range(N):

    # Prediction
    x_pred = A @ x
    P_pred = A @ P @ A.T + G @ Q @ G.T

    # Current measurement
    y = measurements[:, k]

    # Innovation
    v = y - H @ x_pred
    S = H @ P_pred @ H.T + R

    # Kalman gain
    K = P_pred @ H.T @ np.linalg.inv(S)

    # Update
    x = x_pred + K @ v
    P = (np.eye(4) - K @ H) @ P_pred

    filtered[:, k] = x

# Plot results
plt.figure(figsize=(8, 6))

plt.plot(
    true_x[0],
    true_x[1],
    label="True Trajectory",
    linewidth=2
)

plt.scatter(
    measurements[0],
    measurements[1],
    s=15,
    alpha=0.5,
    label="GPS Measurements"
)

plt.plot(
    filtered[0],
    filtered[1],
    linewidth=2,
    label="Kalman Filter"
)

plt.xlabel("X Position")
plt.ylabel("Y Position")
plt.title("2D Position Estimation using Kalman Filter")
plt.grid(True)
plt.legend()
plt.show()
```
