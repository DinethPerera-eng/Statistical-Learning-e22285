# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?


# Answer

The energy eigenvalues of a single quantum harmonic oscillator are

$$
E_j=\hbar\omega\left(j+\frac{1}{2}\right), \qquad j=0,1,2,\ldots
$$

---

## 1. Partition Function

The canonical partition function is

$$
Z(T)=\sum_{j=0}^{\infty}e^{-\frac{E_j}{k_BT}}.
$$

Substituting the energy eigenvalues,

$$
Z(T)=\sum_{j=0}^{\infty}
e^{-\frac{\hbar\omega\left(j+\frac{1}{2}\right)}{k_BT}}.
$$

This can be separated as

$$
Z(T)=
e^{-\frac{\hbar\omega}{2k_BT}}
\sum_{j=0}^{\infty}
\left(e^{-\frac{\hbar\omega}{k_BT}}\right)^j.
$$

This is a geometric series with common ratio

$$
r=e^{-\frac{\hbar\omega}{k_BT}}.
$$

Since

$$
0<e^{-\frac{\hbar\omega}{k_BT}}<1,
$$

the geometric series is convergent.

Using

$$
\sum_{j=0}^{\infty}r^j=\frac{1}{1-r},
$$

we obtain

$$
Z(T)=
\frac{
e^{-\frac{\hbar\omega}{2k_BT}}
}{
1-e^{-\frac{\hbar\omega}{k_BT}}
}.
$$

---

## 2. Thermodynamic Potentials

### Helmholtz Free Energy

The Helmholtz free energy is

$$
F(T)=-k_BT\ln Z.
$$

Using

$$
Z(T)=
\frac{
e^{-\frac{\hbar\omega}{2k_BT}}
}{
1-e^{-\frac{\hbar\omega}{k_BT}}
},
$$

we get

$$
\ln Z=
-\frac{\hbar\omega}{2k_BT}
-\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right).
$$

Therefore,

$$
F(T)=
-k_BT
\left[
-\frac{\hbar\omega}{2k_BT}
-\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right)
\right].
$$

Hence,

$$
F(T)=
\frac{\hbar\omega}{2}
+
k_BT\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right).
$$

---

### Internal Energy

Let

$$
\beta=\frac{1}{k_BT}.
$$

Then,

$$
Z=
\frac{
e^{-\frac{\beta\hbar\omega}{2}}
}{
1-e^{-\beta\hbar\omega}
}.
$$

Therefore,

$$
\ln Z=
-\frac{\beta\hbar\omega}{2}
-\ln\left(1-e^{-\beta\hbar\omega}\right).
$$

The internal energy is

$$
U(T)=
-\frac{\partial \ln Z}{\partial \beta}.
$$

Differentiating,

$$
U(T)=
\hbar\omega
\left[
\frac{1}{2}
+
\frac{1}{e^{\beta\hbar\omega}-1}
\right].
$$

Since

$$
\beta=\frac{1}{k_BT},
$$

we get

$$
U(T)=
\hbar\omega
\left[
\frac{1}{2}
+
\frac{1}{e^{\frac{\hbar\omega}{k_BT}}-1}
\right].
$$

---

## 3. Entropy

The entropy is found using

$$
S(T)=
-\left(\frac{\partial F}{\partial T}\right).
$$

Using

$$
F(T)=
\frac{\hbar\omega}{2}
+
k_BT\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right),
$$

we obtain

$$
S(T)=
-k_B\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right)
+
\frac{\hbar\omega}{T}
\frac{1}{e^{\frac{\hbar\omega}{k_BT}}-1}.
$$

Now,

$$
U=F+TS.
$$

Substituting $F$ and $S$,

$$
\begin{aligned}
F+TS
&=
\frac{\hbar\omega}{2}
+
k_BT\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right) \\
&\quad+
T
\left[
-k_B\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right)
+
\frac{\hbar\omega}{T}
\frac{1}{e^{\frac{\hbar\omega}{k_BT}}-1}
\right].
\end{aligned}
$$

The logarithmic terms cancel. Therefore,

$$
F+TS=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_BT}}-1}.
$$

Hence,

$$
F+TS=
\hbar\omega
\left[
\frac{1}{2}
+
\frac{1}{e^{\frac{\hbar\omega}{k_BT}}-1}
\right].
$$

This is equal to the internal energy $U(T)$. Therefore,

$$
U=F+TS.
$$

---

## 4. Heat Capacity and Hyperbolic Form

The constant-volume heat capacity is

$$
C_V=
\left(\frac{\partial U}{\partial T}\right).
$$

Using

$$
U(T)=
\hbar\omega
\left[
\frac{1}{2}
+
\frac{1}{e^{\frac{\hbar\omega}{k_BT}}-1}
\right],
$$

we get

$$
C_V=
k_B
\left(
\frac{\hbar\omega}{k_BT}
\right)^2
\frac{
e^{\frac{\hbar\omega}{k_BT}}
}{
\left(e^{\frac{\hbar\omega}{k_BT}}-1\right)^2
}.
$$

Now use the hyperbolic identity

$$
\sinh(x)=\frac{e^x-e^{-x}}{2}.
$$

Squaring both sides,

$$
\sinh^2(x)=
\frac{\left(e^x-e^{-x}\right)^2}{4}.
$$

Therefore,

$$
\frac{1}{\sinh^2(x)}
=
\frac{4}{\left(e^x-e^{-x}\right)^2}.
$$

Since

$$
\left(e^x-e^{-x}\right)^2
=
e^{-2x}\left(e^{2x}-1\right)^2,
$$

we get

$$
\frac{1}{\sinh^2(x)}
=
\frac{4e^{2x}}{\left(e^{2x}-1\right)^2}.
$$

Let

$$
x=\frac{\hbar\omega}{2k_BT}.
$$

Then,

$$
2x=\frac{\hbar\omega}{k_BT}.
$$

Therefore,

$$
\frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_BT}\right)}
=
\frac{
4e^{\frac{\hbar\omega}{k_BT}}
}{
\left(e^{\frac{\hbar\omega}{k_BT}}-1\right)^2
}.
$$

Using this identity in the heat capacity expression,

$$
C_V=
k_B
\left(
\frac{\hbar\omega}{2k_BT}
\right)^2
\frac{1}
{
\sinh^2\left(\frac{\hbar\omega}{2k_BT}\right)
}.
$$

---

## 5. High-Temperature Limit

In the high-temperature limit,

$$
k_BT\gg \hbar\omega.
$$

Therefore,

$$
\frac{\hbar\omega}{2k_BT}\ll 1.
$$

For small $x$,

$$
\sinh(x)\approx x.
$$

Using

$$
x=\frac{\hbar\omega}{2k_BT},
$$

the heat capacity becomes

$$
C_V
\approx
k_B
\left(
\frac{\hbar\omega}{2k_BT}
\right)^2
\frac{1}
{
\left(
\frac{\hbar\omega}{2k_BT}
\right)^2
}.
$$

Therefore,

$$
C_V\approx k_B.
$$

Hence, in the high-temperature limit,

$$
C_V\to k_B.
$$

This matches the classical equipartition theorem prediction for a one-dimensional harmonic oscillator.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$


# Answer

Let $X\in(-\infty,\infty)$ and let its probability density function be $p(x)$.

The continuous Shannon entropy is

$$
H[p]=-\int_{-\infty}^{\infty}p(x)\ln p(x)\,dx.
$$

The constraints are

$$
\int_{-\infty}^{\infty}p(x)\,dx=1,
$$

$$
\int_{-\infty}^{\infty}xp(x)\,dx=\mu,
$$

and

$$
\int_{-\infty}^{\infty}(x-\mu)^2p(x)\,dx=\sigma^2.
$$

To maximize entropy subject to these constraints, define the Lagrangian

$$
\begin{aligned}
\mathscr{L}
=
&-\int_{-\infty}^{\infty}p(x)\ln p(x)\,dx \\
&-\lambda_0
\left(
\int_{-\infty}^{\infty}p(x)\,dx-1
\right) \\
&-\lambda_1
\left(
\int_{-\infty}^{\infty}xp(x)\,dx-\mu
\right) \\
&-\lambda_2
\left(
\int_{-\infty}^{\infty}(x-\mu)^2p(x)\,dx-\sigma^2
\right).
\end{aligned}
$$

Taking the functional derivative with respect to $p(x)$ and setting it equal to zero,

$$
\frac{\delta \mathscr{L}}{\delta p(x)}=0.
$$

Therefore,

$$
-\ln p(x)-1-\lambda_0-\lambda_1x-\lambda_2(x-\mu)^2=0.
$$

So,

$$
\ln p(x)
=
-1-\lambda_0-\lambda_1x-\lambda_2(x-\mu)^2.
$$

Taking exponential on both sides,

$$
p(x)
=
e^{-1-\lambda_0-\lambda_1x-\lambda_2(x-\mu)^2}.
$$

This can be written as

$$
p(x)
=
A\exp\left(
-\lambda_1x-\lambda_2(x-\mu)^2
\right),
$$

where

$$
A=e^{-1-\lambda_0}.
$$

Since the distribution is centered at the mean $\mu$, the linear term becomes zero. Hence,

$$
\lambda_1=0.
$$

Therefore,

$$
p(x)
=
A\exp\left(
-\lambda_2(x-\mu)^2
\right).
$$

Using the variance constraint,

$$
\lambda_2=\frac{1}{2\sigma^2}.
$$

Thus,

$$
p(x)
=
A\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right).
$$

Now use the normalization constraint

$$
\int_{-\infty}^{\infty}p(x)\,dx=1.
$$

So,

$$
\int_{-\infty}^{\infty}
A\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right)
dx
=
1.
$$

Using the Gaussian integral,

$$
\int_{-\infty}^{\infty}
\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right)
dx
=
\sqrt{2\pi\sigma^2}.
$$

Therefore,

$$
A\sqrt{2\pi\sigma^2}=1.
$$

Hence,

$$
A=
\frac{1}{\sqrt{2\pi\sigma^2}}.
$$

Substituting $A$ into $p(x)$ gives

$$
p^*(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right).
$$

Therefore, the maximum entropy distribution under fixed mean and variance is the normal distribution

$$
X\sim \mathcal{N}(\mu,\sigma^2).
$$

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



# Answer: Asset Pricing and Portfolio Optimization

Let the future asset price be

$$
S_T>0
$$

and let its continuous probability density function be

$$
p(S_T).
$$

The continuous Shannon entropy is

$$
H_c[p]=-\int_{0}^{\infty}p(S_T)\ln p(S_T)\,dS_T.
$$

The constraints are

$$
\int_{0}^{\infty}p(S_T)\,dS_T=1,
$$

$$
\int_{0}^{\infty}S_Tp(S_T)\,dS_T=102,
$$

$$
\int_{0}^{\infty}S_T^2p(S_T)\,dS_T=10600,
$$

and

$$
\int_{0}^{\infty}\ln(S_T)p(S_T)\,dS_T=4.60.
$$

---

## Maximum Entropy Lagrangian

To maximize entropy subject to the given constraints, define the Lagrangian as

$$
\begin{aligned}
\mathscr{L}
=
&-\int_{0}^{\infty}p(S_T)\ln p(S_T)\,dS_T \\
&-\gamma
\left(
\int_{0}^{\infty}p(S_T)\,dS_T-1
\right) \\
&-\lambda_1
\left(
\int_{0}^{\infty}S_Tp(S_T)\,dS_T-102
\right) \\
&-\lambda_2
\left(
\int_{0}^{\infty}S_T^2p(S_T)\,dS_T-10600
\right) \\
&-\lambda_3
\left(
\int_{0}^{\infty}\ln(S_T)p(S_T)\,dS_T-4.60
\right).
\end{aligned}
$$

Taking the functional derivative with respect to $p(S_T)$ and setting it equal to zero,

$$
\frac{\delta \mathscr{L}}{\delta p(S_T)}=0.
$$

Therefore,

$$
-\ln p(S_T)-1-\gamma-\lambda_1S_T-\lambda_2S_T^2-\lambda_3\ln(S_T)=0.
$$

So,

$$
\ln p(S_T)
=
-1-\gamma-\lambda_1S_T-\lambda_2S_T^2-\lambda_3\ln(S_T).
$$

Taking exponential on both sides,

$$
p(S_T)
=
e^{-1-\gamma}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2-\lambda_3\ln(S_T)
\right).
$$

Using the normalizing constant $Z$,

$$
p^*(S_T)
=
\frac{1}{Z}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2-\lambda_3\ln(S_T)
\right).
$$

Since

$$
\exp\left(-\lambda_3\ln(S_T)\right)=S_T^{-\lambda_3},
$$

the density can be written as

$$
p^*(S_T)
=
\frac{1}{Z}
S_T^{-\lambda_3}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2
\right),
\qquad S_T>0.
$$

---

## Partition Function

The partition function is

$$
Z(\lambda_1,\lambda_2,\lambda_3)
=
\int_{0}^{\infty}
S_T^{-\lambda_3}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2
\right)
\,dS_T.
$$

---

## Constraint Equations

The Lagrange multipliers $\lambda_1,\lambda_2,\lambda_3$ are selected so that the density satisfies the three market constraints.

For the expected price constraint,

$$
\mathbb{E}[S_T]=102.
$$

Therefore,

$$
-\frac{\partial \ln Z}{\partial \lambda_1}=102.
$$

Equivalently,

$$
\frac{1}{Z}
\int_{0}^{\infty}
S_T^{1-\lambda_3}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2
\right)
\,dS_T
=
102.
$$

For the second moment constraint,

$$
\mathbb{E}[S_T^2]=10600.
$$

Therefore,

$$
-\frac{\partial \ln Z}{\partial \lambda_2}=10600.
$$

Equivalently,

$$
\frac{1}{Z}
\int_{0}^{\infty}
S_T^{2-\lambda_3}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2
\right)
\,dS_T
=
10600.
$$

For the logarithmic constraint,

$$
\mathbb{E}[\ln(S_T)]=4.60.
$$

Therefore,

$$
-\frac{\partial \ln Z}{\partial \lambda_3}=4.60.
$$

Equivalently,

$$
\frac{1}{Z}
\int_{0}^{\infty}
\ln(S_T)
S_T^{-\lambda_3}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2
\right)
\,dS_T
=
4.60.
$$

---

## Final Maximum Entropy Density

Thus, the Maximum Entropy risk-neutral asset price distribution is

$$
p^*(S_T)
=
\frac{1}{Z}
S_T^{-\lambda_3}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2
\right),
\qquad S_T>0.
$$

where

$$
Z=
\int_{0}^{\infty}
S_T^{-\lambda_3}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2
\right)
\,dS_T.
$$

The parameters $\lambda_1,\lambda_2,\lambda_3$ are determined by

$$
\mathbb{E}[S_T]=102,
$$

$$
\mathbb{E}[S_T^2]=10600,
$$

and

$$
\mathbb{E}[\ln(S_T)]=4.60.
$$

Therefore, the required continuous probability density function is

$$
p^*(S_T)
=
\frac{1}{Z}
S_T^{-\lambda_3}
\exp\left(
-\lambda_1S_T-\lambda_2S_T^2
\right),
\qquad S_T>0.
$$
